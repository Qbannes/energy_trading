# Python Debugger

Du hast wahrscheinlich eine Menge von `print`-Anweisungen verwendet, um zu versuchen, die Errors in deinem Code zu finden. Ein besserer Weg dazu, ist die Verwendung des vorinstallierten (built-in) Debugger Moduls `pdb`. Das Modul `pdb` implementiert eine interaktive Debugging Umgebung für Python-Programme. Es beinhaltet Features, um dein Programm zu pausieren, die Werte von Variablen anzuschauen und die Programmausführung Schritt für Schritt zu analysieren. So kannst du verstehen, was dein Programm wirklich macht und Fehler in der Logik finden.

Das zu zeigen ist ein wenig schwierig, da wir absichtlich einen Error erzeugen müssen. Hoffentlich verdeutlicht dieses einfache Beispiel den großen Nutzen des Moduls `pdb`.

*Beachtet: Die Nutzung in einem iPython Notebook Setting ist nicht der normale Anwendungsfall.*

Hier erzeugen wir einen Fehler, indem wir versuchen, eine Liste zu einer Integer dazuzuaddieren.

```python
x = [1, 3, 4]
y = 2
z = 3

ergebnis = y + z
print(ergebnis)
ergebnis2 = y+x
print(ergebnis2)
```

Hmmm, wo kommt bloß dieser Error her? Lasst uns ein `set_trace` (etwa: Spur setzen) einbinden, das im Modul `pdb` enthalten ist. Das erlaubt uns, den Code an der Stelle der Anweisung anzuhalten.

```python
import pdb

x = [1, 3, 4]
y = 2
z = 3

ergebnis = y + z
print(ergebnis)

# Eine Spur mit dem pdb setzen
pdb.set_trace()

ergebnis2 = y+x
print(ergebnis2)
```

So können wir die verschiedenen Variablen im Code überprüfen und auf Fehler kontrollieren. Mit `c` (continue, fortfahren) wird die Programmausführung fortgesetzt. Du kannst `q` (quit, beenden) benutzen, um den Debugger zu beenden. Für weitere Informationen könnt ihr euch die [Dokumentation](https://docs.python.org/3/library/pdb.html) anschauen.
