# Code-Laufzeit messen

Manchmal ist es wichtig, zu wissen, wie lange dein Code zur Ausführung benötigt oder wenigstens, ob eine bestimmte Zeile den Code verlangsamt. Python verfügt über Modul `timeit`, um genau das herauszufinden.

## Beispielfunktion

Hier haben wir zwei Funktionen, die das gleiche leisten, aber auf verschiedene Arten. Welche dieser Funktionen ist effizienter? Lasst uns das herausfinden!

```python
def func_eins(n):
    """ Gibt eine Liste aller Integers bis n als String zurück. """

    return [str(num) for num in range(n)]
```

```python
func_eins(10)
```

```python
def func_zwei(n):
    """ Gibt eine Liste aller Integers bis n als String zurück. """

    return list(map(str, range(n)))
```

```python
func_zwei(10)
```

### Start und Stop messen

Wir können versuchen, das Modul `time` zu verwenden, um die Laufzeit des Codes zu messen. Dabei ist aber die Präzision von `time` zu beachten - der Code muss **mindestens** eine Laufzeit von 0.1 Sekunden haben!

```python
import time
```

```python
# Schritt 1: Startzeit ermitteln
startzeit = time.time()
# Schritt 2: Code ausführen
ergebnis = func_eins(1000000)
# Schritt 3: Vergangene Zeit berechnen
laufzeit = time.time() - startzeit
```

```python
laufzeit
```

```python
# Schritt 1: Startzeit ermitteln
startzeit = time.time()
# Schritt 2: Code ausführen
ergebnis = func_zwei(1000000)
# Schritt 3: Vergangene Zeit berechnen
laufzeit = time.time() - startzeit
```

```python
laufzeit
```

### Modul `timeit`

Und was machen wir, wenn wir Code messen wollen, der zu schnell läuft für eine Messung mit `time.time()`? Dann setzen wir das Modul `timeit` ein.

`timeit` nimmt drei Parameter an: Statement (etwa: Ausdruck), welches den Code zum ausführen des Testcodes enthält, den Testcode (Setup) und zuletzt die Anzahl der Ausführungen. Es wird dann die Gesamtlaufzeit zurückgegeben. 

```python
import timeit
```

Das Setup (Alles, was vor Ausführung definiert werden muss, Funktionen, Variablen, etc.)

```python
setup = '''
def func_eins(n):
    return [str(num) for num in range(n)]
'''
```

```python
stmt = 'func_eins(100)'
```

```python
timeit.timeit(stmt, setup, number=100000)
```

Jetzt lassen wir auch `func_zwei` 10.000 mal laufen und vergleichen die Laufzeit.

```python
setup2 = '''
def func_zwei(n):
    return list(map(str, range(n)))
'''
```

```python
stmt2 = 'func_zwei(100)'
```

```python
timeit.timeit(stmt2, setup2, number=100000)
```

Scheint so, als sei `func_zwei` effizienter. Wir können das noch genauer untersuchen, indem wir die Anzahl von Durchläufen erhöhen.

```python
timeit.timeit(stmt, setup, number=1000000)
```

```python
timeit.timeit(stmt2, setup2, number=1000000)
```

## Code-Laufzeit mit Jupyter Magic messen

**Beachte: Diese Methode funktioniert NUR in Jupyter und die Magic muss zu Beginn der Zelle mit nichts darüber stehen (nicht einmal Kommentare).**

```python
%%timeit
func_eins(100)
```

```python
%%timeit
func_zwei(100)
```
