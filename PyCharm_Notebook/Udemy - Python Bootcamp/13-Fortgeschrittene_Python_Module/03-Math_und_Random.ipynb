# Modul Random

Das Modul `random` erlaubt es uns, Zufallszahlen zu erzeugen. Wir können sogar einen Seed (Samen) erzeugen, wodurch jedesmal der gleiche Satz Zufallszahlen erzeugt wird.

Die Erklärung, wie ein Computer versucht, Zufallszahlen zu erzeugen, ist ausserhalb der Möglichkeiten dieses Kurses, da dazu anspruchsvolle Mathematik erforderlich ist. Aber wenn dich das Thema interessiert, dann wird mal einen Blick auf diese Quellen:
* https://en.wikipedia.org/wiki/Pseudorandom_number_generator
* https://en.wikipedia.org/wiki/Random_seed

## Seed verstehen

Durch das setzen eines Seeds setzen wir den Startwert des Pseudozufallszahlengenerators, wodurch die gleiche Serie von Zufallszahlen erzeugt wird. In Jupyter musst du den Seed in der gleichen Zelle setzen, damit auch wirklich die gleichen Zufallszahlen erzeugt werden. Der Zugriff auf den gleichen Satz von Zufallszahlen ist dann wichtig, wenn du die Leistung von Funktionen auf Zufallszahlen testen und dabei den Test fair halten willst (wofür die gleichen Zufallszahlen nötig sind). 

```python
import random
```

```python
random.randint(0, 100)
```

```python
random.randint(0, 100)
```

```python
# Der Wert 101 ist beliegig, du kannst irgendeine Zahl wählen
random.seed(101)

# Du kannst diese Zelle so oft du willst ausführen, das Ergebnis bleibt gleich
random.randint(0, 100)
```

```python
random.randint(0, 100)
```

```python
random.seed(101)
print(random.randint(0,100))
print(random.randint(0,100))
print(random.randint(0,100))
print(random.randint(0,100))
print(random.randint(0,100))
```

### Samplen mit Ersatz

Wähle eine Sample-Größe und erlaube die wiederholte Auswahl eines Elements. Stelle dir einen Beutel mit Losen vor, die nach jedem ziehen **zurück in den Beutel gelegt werden**, bevor das nächste Los gezogen wird.

```python
random.choices(population=meine_liste, k=10)
```

### Sample ohne Ersatz

Wenn ein Element gewählt wurde, kann es nicht erneut gewählt werden. Stelle dir einen Beutel mit Losen vor, bei dem bei jedem ziehen **das Los entfernt wird**, bevor das nächste Los gezogen wird.

```python
random.sample(population=meine_liste, k=10)
```

### Eine Liste mischen

**Beachte: Dies mischt die Liste in-place!**

```python
# Diese Methode gibt kein Ergebnis zurück und wird daher nicht zugewiesen!
random.shuffle(meine_liste)
```

```python
meine_liste
```

### Zufallsverteilungen

#### [Uniforme Verteilung](https://en.wikipedia.org/wiki/Uniform_distribution)

```python
# Kontinuierliche, zufällige Wahl eines Wertes zwischen a und b, jeder Wert
# hat die gleiche Chance, gewählt zu werden
random.uniform(a=0, b=100)
```

#### [Normal/Gaussian Distribution](https://en.wikipedia.org/wiki/Normal_distribution)

```python
random.gauss(mu=0, sigma=1)
```
