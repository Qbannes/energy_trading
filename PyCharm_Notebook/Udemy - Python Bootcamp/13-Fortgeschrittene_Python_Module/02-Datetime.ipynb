# Datetime

Python hat das `datetime` Modul integriert, um dabei zu helfen, in eurem Code mit Zeitstempeln zu arbeiten. Zeitwerte werden durch die Klasse `time` repräsentiert. Zeiten haben Attribute für Stunden, Minuten, Sekunden und Mikrosekunden. Sie können außerdem Zeitzoneninformation beinhalten. Die Parameter, mit der eine Instanz von `time` initialisiert wird, sind optional und per Standard auf 0 gesetzt. Das wird aber kaum die Zeit sein, die du haben möchtest.

## time

Lasst uns betrachten, wie wir Zeitinformationen aus dem `datetime` Modul auslesen. Wir können einen Zeitstempel erstellen, indem wir `datetime.time(hour, minute, second, microsecond)` aufrufen.

```python
import datetime

zeit = datetime.time(4, 20, 1)

# Lasst uns die Bestandteile betrachten
print(zeit)
print('Stunde:', zeit.hour)
print('Minute:', zeit.minute)
print('Sekunde:', zeit.second)
print('Mikrosekunde:', zeit.microsecond)
print('Zeitzone:', zeit.tzinfo)
```

**Beachte:** eine `time`-Instanz enthält nur Werte für Zeit, nicht jedoch für das Datum. 

Wir können uns auch die Mindest- und Höchstwerte für Zeitwerte anschauen:

```python
print('Mindestwert:', datetime.time.min)
print('Höchstwert:', datetime.time.max)
print('Schrittgröße:', datetime.time.resolution)
```

Die Klassenattribute `min` und `max` geben die maximalen Zeitwerte innerhalb eines einzigen Tages zurück.

## Daten

`datetime` (wie du es dir bestimmt schon gedacht hast) erlaubt es uns auch, mit Datumsstempeln zu arbeiten. Kalenderdaten werden durch die Klasse `date` repräsentiert. Sie besitzen Parameter für das Jahr, den Monat und den Tag. Es ist allerdings einfacher, ein Datum zu erstellen, dass den heutigen Tag wiedergibt, indem wir `today` verwenden.

Hier einige Beispiele:

```python
heute = datetime.date.today()
print(heute)
print('Weltzeit', heute.ctime())
print('Tupel:', heute.timetuple())
print('Ordinal:', heute.toordinal())
print('Jahr:', heute.year)
print('Monat :', heute.month)
print('Tag :', heute.day)
```

Genau wie bei `time` können wir uns Mindest- und Höchstwerte ausgeben lassen, indem wir `min` und `max` verwenden.

```python
print('Mindestwert:',datetime.date.min)
print('Höchstwert:',datetime.date.max)
print('Schrittgröße:',datetime.date.resolution)
```

Ein anderer Weg, um neue Datuminstanzen zu erzeugen, ist die Verwendung der Methode `replace` mit einem bestehenden Datum. Zum Beispiel können wir das Jahr ändern und Tag und Monat unberührt lassen.

```python
datum1 = datetime.date(2015, 3, 11)
print('datum1:', datum1)

datum2 = datum1.replace(year=1990)
print('datum2:', datum2)
```

## Arithmetik

Wir können mit auch Arithmetik auf `date`-Instanzen anwenden:

```python
datum1
```

```python
datum2
```

```python
datum1 - datum2
```

Wir erhalten die Differenz zwischen den beiden Daten in Tagen. Wir können auch die Methode `timedelta` verwenden, um verschiedenste Einheiten von Zeit zu spezifizieren (Tage, Minuten, Stunden,...).
