# Dateien packen und entpacken

Dir ist vermutlich bekannt, dass Dateien im Zip-Format komprimiert werden können. Häufig werden dazu spezielle Programme verwendet. Erfreulicherweise kann Python das auch, mit nur wenigen Zeilen Programmcode. 

## Dateien zum packen erzeugen

```python
datei = open("neue_datei.txt", 'w+')
datei.write("Hier ist Text")
datei.close()
```

```python
datei = open("neue_datei2.txt", 'w+')
datei.write("Hier ist Text")
datei.close()
```

## Dateien packen

Das Packet [zipfile](https://docs.python.org/3/library/zipfile.html) ist Teil der Standardbibliothek von Python. Wir können es zum packen und entpacken von Dateien und Verzeichnissen nutzen. Um alle Dateien in einem Verzeichnis zu packen, kannst du die vorher schon beschriebene Methode `os.walk` einsetzen, um durch die Verzeichnisstruktur zu iterieren.

```python
import zipfile
```

Erzeuge zuerst die Zip-Datei und schreibe anschliessend in diese (der Schreib-Schritt komprimiert die Dateien).

```python
komp_datei = zipfile.ZipFile('komp_datei.zip', 'w')
```

```python
komp_datei.write("neue_datei.txt", compress_type=zipfile.ZIP_DEFLATED)
```

```python
komp_datei.write('neue_datei2.txt', compress_type=zipfile.ZIP_DEFLATED)
```

```python
komp_datei.close()
```

## Zip-Dateien entpacken

Wir können Datien nun einfach entpacken, entweder mit der Methode `.extractall` für alle Dateien gleichzeitig oder mit `.extract()` für einzelne Dateien.

```python
zip_obj = zipfile.ZipFile('komp_datei.zip','r')
```

```python
zip_obj.extractall("zip_entpackt")
```
