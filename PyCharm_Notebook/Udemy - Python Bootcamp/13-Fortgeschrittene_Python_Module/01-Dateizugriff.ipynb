### Dateien verschieben

Du kannst das built-in-Modul **shutil** nutzen, um Dateien an andere Orte zu verschieben. Dabei musst du allerdings Zugriffsrechte beachten: Wenn du als Nutzer A eingeloggt bist, kannst du beispielsweise nicht an Orte verschieben, auf die nur der Administrator Zugriff hat, [siehe auch hier](https://stackoverflow.com/questions/23253439/shutil-movescr-dst-gets-me-ioerror-errno-13-permission-denied-and-3-more-e) 

```python
import shutil
```

```python
shutil.move('testdatei.txt','/home/anton')
```

```python
os.listdir()
```

```python
shutil.move('/home/anton/testdatei.txt',os.getcwd())
```

```python
os.listdir()
```

### Dateien löschen
____
**Beachte: Das Modul `os` bietet drei Methoden zum löschen von Dateien:**
* os.unlink(pfad) löscht die Datei in pfad
* os.rmdir(pfad) löscht das **leere** Verzeichnis in Pfad
* shutil.rmtree(pfad) löscht alle in pfad enthaltenen Dateien und Verzeichnisse (**gefährlich**)

**Alle diese Methoden können nicht rückgängig gemacht werden! Das heisst, wenn du einen Fehler machst, kann die Datei nicht wiederhergestellt werden. Stattdessen verwenden wir das Modul `send2trash`, eine sicherere Alternative, welche Dateien in den Papierkorb schickt, anstatt sie permanent zu löschen.**
___

Installiere das Modul `send2trash` mit:

    pip install send2trash
    
in der Kommandozeile.

```python
import send2trash
```

```python
os.listdir()
```

```python
send2trash.send2trash('testdatei.txt')
```

```python
os.listdir()
```

### Durch ein Verzeichnis iterieren

In vielen fällen wirst du durch ein Verzeichnis iterieren wollen und dabei alle enthaltenen Dateien und Verzeichnisse betrachten, zum Beispiel um zu sehen, ob eine Datei vorhanden ist, um sie dann ggf. zu bearbeiten. Eine rekursive Iteration auch durch Unterverzeichnisse kann schwierig zu programmieren sein, aber glücklicherweise bietet das Modul `os` direkt eine Methode namens `os.walk` dafür an. Lasst uns anschauen, wie das funktioniert.

```python
os.getcwd()
```

```python
os.listdir()
```

```python
for verzeichnis, unterverzeichnisse, dateien in os.walk("Beispiel_Top_Level"):

    print("Du betrachtest gerade Verzeichnis: " + verzeichnis)
    print('\n')
    print("Die Unterverzeichnisse sind: ")
    for unter in unterverzeichnisse:
        print("\t Unterverzeichnis: " + unter)

    print('\n')

    print("Die Dateien sind: ")
    for datei in dateien:
        print("\t Datei: " + datei)
    print('\n')

    # Wirf einen Blick auf die Verzeichnisstruktur für ein besseres Verständnis!
```

___
Du solltest jetzt wissen, wie du mit Dateien und Verzeichnissen an beliebigen Speicherorten arbeitest. Sympathischerweise funktioniert das Modul `os` mit jedem Betriebssystem, das Python unterstützt, also Linux, MacOS und Windows, und das ohne jegliche Änderungen!
