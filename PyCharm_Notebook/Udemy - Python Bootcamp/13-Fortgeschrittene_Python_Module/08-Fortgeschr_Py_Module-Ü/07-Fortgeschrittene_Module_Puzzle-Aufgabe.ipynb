# Fortgeschrittene Module Puzzle - Aufgabe

Es ist an der Zeit, deine neuen Fertigkeiten auszuprobieren. Dieses Puzzle kombiniert mehrere der in diesem Abschnitt gelernten Inhalte, inklusive dem Entpacken von Dateien und der Verwendung des Modules `os`, um durch grosse Mengen von Dateien zu iterieren.

## Dein Ziel

Dies ist ein Puzzle, von daher wirst du nur ein Minimum an Anleitung bekommen und deinen Lösungsweg selbst suchen müssen.

Im Verzeichnis liegt eine Datei "entpacke_mich.zip". Entpacke diese Datei, öffne die enthaltene .txt-Datei mit Python, lies die Anleitung und schau, ob du herausfinden kannst, was zu tun ist!
