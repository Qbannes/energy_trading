## Schritt 3: Reguläre Ausdrücke zum suchen der Telephonnummer

Hier gibt es viele Ansätze, aber da wir nach einer Telephonnummer suchen, sollten wir versuchen, das Muster ###-###-#### zu finden. Dafür können wir einfach einen regulären Ausdruck verwenden. Am besten testen wir diesen Ausdruck zuerst und wenden ihn anschliessend auf alle Textdateien an.

```python
import re
```

```python
muster = r'\d{3}-\d{3}-\d{4}'
```

```python
test_string = "hier ist irgendeine Zahl 1231231234 und hier eine Telephonnummer im Format 123-123-1234."
```

```python
re.findall(muster, test_string)
```

## Schritt 4: Regex-Funktion zur Suche in einer Datei

Lasst uns die Regex-Suche in eine Funktion packen, die einen Pfad als Parameter annimmt und in der entsprechenden Datei nach dem Muster sucht. Diese Funktion können wir dann auf alle Dateien im Verzeichnis "entpackter_inhalt" anwenden.

```python
def suche(datei, muster=r'\d{3}-\d{3}-\d{4}'):
    f = open(datei, 'r')
    text = f.read()

    if re.search(muster, text):
        return re.search(muster, text)
    else:
        return ''
```

## Schritt 5: Iteriere mit dem Modul `os` über alle Dateien und suche das Muster

Nachdem wir jetzt eine einfache Funktion für die Suche nach dem Muster haben, können wir mit `os.walk` über die Dateien iterieren und nach der Telephonnummer suchen.

```python
import os
```

```python
ergebnisse = []

for verzeichnis, unter_verz, dateien in os.walk(os.path.join(os.getcwd(), "entpackter_inhalt")):

    for f in dateien:
        ganzer_pfad = os.path.join(verzeichnis, f)

        ergebnisse.append(suche(ganzer_pfad))
```

```python
for ergebnis in ergebnisse:
    if ergebnis != '':
        print(ergebnis.group())
```
