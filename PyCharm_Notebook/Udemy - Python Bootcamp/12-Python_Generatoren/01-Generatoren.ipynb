# Iteratoren und Generatoren

In diesem Abschnitt des Kurses werden wir mehr über die Unterschiede von Iteration und Generation in Python lernen. Und darüber hinaus, wie wir durch die `yield` Anweisung unsere eigenen Generatoren erstellen können. Generatoren erlauben es uns, Schritt-für-Schritt Ergebnisse zu generieren, anstatt alle Ergebnisse gleichzeitig im Speicher zu halten.

Wir haben das bereits thematisiert, zum Beispiel als wir uns die built-in-Funktionen `range`, `map` und `filter` angeschaut haben. 

Wir können das ein wenig vertiefen. Wir wissen bereits, wie wir mit `def` und `return` Funktionen erstellen können. Generator-Funktionen erlauben es uns, eine Funktion zu schreiben, die einen Wert ausgibt und später dort weitermacht, wo sie aufgehört hat. Diese Art von Funktion nennt man in Python *Generator*. Der Hauptunterschied in der Syntax ist die Anweisung `yield` statt `return` zur Rückgabe eines Ergebnisses.

In den meisten Aspekten wirken Generatoren wie normale Funktionen. Sie unterscheiden sich hauptsächlich darin, dass eine Generator-Funktion, sobald sie ausgeführt wird, zu einem Objekt wird, welches das Iterationsprotokoll unterstützt. Das bedeutet, dass sie beim Ausführen nicht nur einen Wert zurückgeben und dann beenden. Die Generatorfunktion pausiert automatisch, bleibt aber an dieser Stelle. Der größte Vorteil davon liegt darin, dass nicht immer alle Werte von Anfang an berechnet werden müssen, sondern die Generatorfunktion nach jedem Rechenschritt wartet. Dieses Feature nennt man *state suspension*. 

Um ein besseres Verständnis für Generatoren zu erhalten, lasst uns einige erstellen:

```python
def gen_raum(grenze):
    """ Generatorfunktion für die Rauminhalte (hoch 3) bis zu grenze. """

    for num in range(grenze):
        yield num**3
```

```python
for x in gen_raum(10):
    print(x)
```

Wunderbar! Jetzt müssen wir dank unserer Generatorfunktion nicht mehr jeden einzelnen Rauminhalt speichern, den wir eventuell einmal benötigen. 

Generatoren sind am besten dafür geeignet, große Mengen von Ergebnissen zu berechnen. Und das in Fällen, in denen wir keinen Speicher für alle Ergebnisse auf einmal aufbringen wollen.

Lasst uns als weiteres Beispiel [Fibonaccireihen](https://en.wikipedia.org/wiki/Fibonacci_number) wählen:

```python
def gen_fibon(grenze):
    """ Generiere alle Fibonnaci-Zahlen bis grenze. """
    
    a = 1
    b = 1
    
    for i in range(grenze):
        yield a
        a, b = b, a + b
```

```python
for num in gen_fibon(10):
    print(num)
```

Wie würde dies als normale Funktion aussehen?

```python
def fibonacci(grenze):
    """ Generiere alle Fibonnaci-Zahlen bis grenze. """

    a = 1
    b = 1
    ergebnis = []

    for i in range(grenze):
        ergebnis.append(a)
        a, b = b, a + b

    return ergebnis
```

```python
fibonacci(10)
```

Beachte, dass wenn wir einen großen Wert für n (z.B. 100000) eingeben würden, die zweite Funktion alle einzelnen Ergebnisse speichern müsste. Dabei brauchen wir eigentlich immer nur das vorherige Ergebnis, um das nächste zu berechnen!

## next und iter

Die beiden buit-in-Funktionen `next` und `iter` sind ein wichtiger Schlüssel zum vollen Verständnis von Generatoren.

Die `next` Funktion erlaubt es uns, das nächste (next) Element einer Sequenz zu wählen. Siehe her:

```python
def einfacher_gen():
    """ Einfacher Generator """
    
    for x in range(3):
        yield x
```

```python
generator = einfacher_gen()
```

```python
print(next(generator))
```

```python
print(next(generator))
```

```python
print(next(generator))
```

```python
print(next(generator))
```

Nachdem wir alle Werte "entnommen" haben, führt `next` zu einem *Stopiteration*-Error. Worüber uns dieser Error informiert, ist die Tatsache, dass alle Werte erzeugt wurden.

Du könntest dich wundern, warum wir diesen Error nicht bei Schleifen erhalten? Die `for`-Schleife achtet selbst auf diesen Error und stoppt dann die Iteration. 

Lasst uns weitergehen und sehen, wie wir `iter` verwenden. Du weisst sicherlich noch, dass Strings iterierbar sind:

```python
mein_string = 'hallo'


for zeichen in mein_string:
    print(zeichen)
```

Das bedeutet allerdings nicht, dass Strings selbst ein Iterator sind. Wir können das mit der Funktion `next` überprüfen:

```python
next(mein_string)
```

Interessant! Das bedeutet, dass String-Objekte Iteration unterstützen, wir aber nicht direkt darüber iterieren können, wie es eine Generatorfunktion könnte. Die Funktion `iter` Funktion erlaubt uns genau das!

```python
s_iter = iter(mein_string)
```

```python
next(s_iter)
```

```python
next(s_iter)
```

Jetzt wisst ihr, wie man Objekte, die iterierbar sind, in Iteratoren umwandelt!

Der wichtigste Teil dieser Lektion ist es, die Anweisung `yield` einzusetzen, um aus einer Funktion einen Generator zu machen. Diese Änderung kann euch eine Menge Speicherkapazität in Anwendungsfällen sparen.
