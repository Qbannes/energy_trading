## Erweitern

Oft ist es zu Beginn etwas schwer, den Unterschied zwischen anhängen und erweitern zu verstehen. Also merke dir:

**append: ein Element am Ende anhängen**

```python
list1 = [1, 2, 3]
list1.append([4, 5])
print(list1)
```

**extend: alle Elemente eines iterierbaren Objektes an die Liste anhängen:**

```python
list1 = [1, 2, 3]
list1.extend([4, 5])
print(list1)
```

Beachte, wie `.extend` jedes einzelne Element (nacheinander) anhängt und `.append` nur das eine Element (hier eine Liste). Das ist der Hauptunterschied.

# Pass auf bei Zuweisungen!

Ein häufiger Programmierfehler ist die Annahme, dass man eine Liste ohne weiteres einer neuen Variablen zuweisen kann. Das funktioniert mit immutablen Objekten wie Strings und Tupeln, aber nicht bei Listen:

```python
mein_string = 'Hallo Welt'
```

```python
mein_string2 = mein_string.upper()
```

```python
print(mein_string2)
```

Das klappt so mit Listen **nicht**:

```python
liste = [1,2,3]
```

```python
liste2 = liste.append(4)
```

```python
print(liste2)
```

Was ist passiert? Die Listenmethode `.append` modifiziert die Liste in-place und die Methode gibt `None` zurück, was an `liste2` übergeben wurde. Statt dessen muss `liste2` eine Kopie von `liste` zugewiesen und `liste2` anschliessend modifiziert werden:

```python
liste = [1, 2, 3]
liste2 = liste.copy()
liste2.append(4)
```

```python
print(liste)
```

```python
print(liste2)
```
