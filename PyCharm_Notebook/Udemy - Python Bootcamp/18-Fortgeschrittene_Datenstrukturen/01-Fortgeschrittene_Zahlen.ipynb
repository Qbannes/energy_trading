# Fortgeschrittene Zahlen

In dieser Lektion werden wir mehr über die Darstellung von Zahlen in Python lernen.

## Hexadezimal

Durch die Nutzung der Funktion hex() können wir Zahlen in das Hexadeziamlformat [Hexadaziamlformat](https://de.wikipedia.org/wiki/Hexadezimalsystem) umwandeln:

```python
hex(246)
```

```python
hex(512)
```

## Binär

Die Funktion `bin` wandelt Zahlen in das Binärformat [Binärformat](https://de.wikipedia.org/wiki/Dualsystem) um:

```python
bin(1234)
```

```python
bin(128)
```

```python
bin(512)
```

## Exponenieren

Exponieren erfolgt mit der Funktion `pow` mit zwei Parametern `x` und `y` und dem Ergebnis `x^y`. Mit einem dritten Parameter `z` ist das Ergebnis `(x^y)%z`, aber effizienter für grosse Integers.

```python
pow(3, 4)
```

```python
pow(3, 4, 5)
```

## Absoluter Wert
Mit der Funktion `abs` erhältst du den absoluten Wert einer Zahl. Je nach Eingabe ist das Ergebnis ein Integer, ein Float, oder, im Falle einer komplexen Zahl, deren Größenordnung.

```python
abs(-3.14)
```

```python
abs(3)
```

## Runden

`round` rundet die Eingabezahl auf die Nachkommastellenzahl des zweiten Eingabeparameters (default ist 0). Dabei werden Integers nicht zu Floats konveriert.

```python
round(3, 2)
```

```python
round(395, -2)
```

```python
round(3.1415926535, 2)
```
