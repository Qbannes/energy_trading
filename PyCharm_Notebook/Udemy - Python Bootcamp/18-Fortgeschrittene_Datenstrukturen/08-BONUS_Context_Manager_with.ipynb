# Context Manager with

Wenn du mit `f = open('test.txt')` eine Datei öffnest, bleibt die Datei offen, bis du explizit `f.close()` aufrufst. Sollte beim bearbeiten der Datei eine Exception auftreten, so bleibt die Datei geöffnet. Das ist sowohl ineffizient als auch angreifbar.

Mit einem Context Manager wird das Öffnen und Schließen von Ressourcen kontrolliert. Dazu gehört ein eingebauter `try/finally`-Block, falls eine Exception auftreten sollte.

Das lässt sich am bestem mit einem Beispiel verdeutlichen.


### Standardprozedur `open()` mit Exception:


```python
f = open('oops.txt','a')
f.readlines()
f.close()
```

Können wir trotzdem in die Datei schreiben?

```python
f.write('mehr text')
```

Autsch! Das hätte ich vielleicht erst tun wollen, nachdem ich die Exception untersucht und gelöst habe. Unglücklicherweise hat die Exception die Ausführung der Zeile `f.close()` verhindert. Lasst uns die Datei manuell schliessen:

```python
f.close()
```

### Schütze die Datei mit `try/except/finally`

Eine gebräuchliche Lösung ist das Einfügen eines `try/except/finally`-Blocks, umd die Datei sicher zu schliessen, falls eine Exception auftritt. 

```python
f = open('oops.txt','a')

try:
    f.readlines()
except:
    print('Eine Exception ist aufgetreten!')
finally:
    f.close()
```

Können wir die Datei noch modifizieren?

```python
f.write('füge Text hinzu')
```

Super, unsere Datei ist sicher!

### Sicherer Zugriff mit `with`

Jetzt setzen wir einen Context Manager ein. Die Syntax entspricht `with [Ressource] as [Name]: mache etwas`

```python
with open('oops.txt','a') as f:
    f.readlines()
```


Können wir die Datei modifizieren?

```python
f.write('füge Text hinzu')
```

Spitze! Mit nur einer Zeile Code haben wir die Datei geöffnet, in einem `try/finally`-Block eingeschlossen und nach Bearbeitung wieder geschlossen.

Jetzt solltest du die Grundlagen von Context Managers verstanden haben.
