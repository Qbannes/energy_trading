# Fortgeschrittene Strings

String-Objekte bieten eine Vielzahl an interessanten Methoden, die uns viel Programmierzeit sparen können. In dieser Lektion werden wir einige davon ausprobieren:

```python
mein_str = "hallo Welt"
```

## Groß-/Kleinschreibung

Wir können Methoden nutzen, um den ersten Buchstaben groß zu schreiben oder alle Wörter eines Strings klein bzw. groß zu schreiben.

```python
# Den ersten Buchstaben eines Strings groß schreiben
mein_str.capitalize()
```

```python
mein_str.upper()
```

```python
mein_str.lower()
```

Bedenke dabei, dass Strings immutable sind. Keine der obigen Methoden ändert den String an sich, sondern sie geben modifizierte Versionen des Originalstrings zurück:

```python
mein_str
```

Zum ändern des Strings ist also eine Neuzuweisung erforderlich:

```python
mein_str = mein_str.upper()
mein_str
```

```python
mein_str = mein_str.lower()
mein_str
```

## Zählen und suchen

```python
# Gibt die Anzahl der Zeichenfolge im String zurück
mein_str.count('l') 
```

```python
# Gibt den Index des ersten Auftretens der Zeichenfolge zurück
mein_str.find('o')
```

## Formatierung

Die Methode `.center` gestattet es, unseren String in der Mitte innerhalb eines Strings mit einer bestimmten Länge zu platzieren. Aber um ehrlich zu sein, kam diese ein bisschen esotherische Methode in keinem meiner Codes bisher zum Einsatz...

```python
mein_str.center(20, 'z')
```

Die Methode `.expandtabs` ersetzt den Tabulator `\t` durch Leerzeichen:

```python
'hallo\thi'.expandtabs()
```

## Methoden zur Typbestimmung

Diese folgenden Methoden überprüfen, ob der String bestimmte Bedingungen erfüllt. Schauen wir sie uns an:

```python
mein_str = 'hallo'
```

`.isalnum` gibt `True` aus, wenn alle Zeichen im String alphanumerisch sind.

```python
mein_str.isalnum()
```

`.isalpha` gibt `True` zurück, falls alle Zeichen im String Buchstaben des Alphabets sind.

```python
mein_str.isalpha()
```

`.islower` gibt `True` aus, falls alle Zeichen im String klein geschrieben sind und der String mindestens eine Länge von 1 hat. Andernfalls erhalten wir `False`.

```python
mein_str.islower()
```

`.isspace()` gibt `True` zurück, wenn alle Zeichen im String Leerzeichen sind.

```python
mein_str.isspace()
```

Mit `.istitle` testen wir, ob der Eingabestring mindestens ein Zeichen hat und einem grossgeschriebenem Wort entspricht, also der erste Buchstabe gross- und der Rest des Strings kleingeschrieben ist.

```python
mein_str.istitle()
```

`.isupper` gibt `True` aus, falls alle Zeichen im String groß geschrieben sind, sonst `False`.

```python
mein_str.isupper()
```

`.endswith` überprüft, ob das letzte Zeichen des Strings dem übergebenen Parameter entspricht.

```python
mein_str.endswith('o')
```

## Eingebaute reguläre Ausdrücke

Strings haben einige vorinstallierte Methoden, die Operationen mit regulären Ausdrücken ausführen. Wir können `.split` benutzen, um den String an einem bestimmten Element zu trennen und das Ergebnis in einer Liste auszugeben.

Wir können außerdem `.partition` verwenden, um ein Tupel zu erzeugen, das den Teiler (erste Erscheinung) sowie die vordere und hintere Hälfte beinhaltet.

```python
mein_str.split('a')
```

```python
mein_str.partition('l')
```
