# Fortgeschrittene Sets

In dieser Lektion werden wir uns mit verschiedenen Methoden für Sets auseinandersetzen, die du bis jetzt noch nicht gesehen haben könntest. Wir wiederholen die bekannten und tauchen dann ein wenig tiefer ein. Los geht's!

```python
mein_set = set()
```

## Hinzufügen

Um Elemente zum Set hinzuzufügen nutzen wir `.add`. Denke daran, dass ein Set keine duplizierten Elemente aufnimmt und sie nur einmal darstellt. Deshalb ist es ein Set!

```python
mein_set.add(1)
```

```python
mein_set.add(2)
```

```python
mein_set.add(2)
```

```python
mein_set
```

## Zurücksetzen

`.clear` entfernt alle Elemente aus dem Set.

```python
mein_set.clear()
```

```python
mein_set
```

## Kopieren

Die Methode `.copy` kopiert ein Set. Beachte, dass es wirkliche eine Kopie ist, d.h. Veränderungen an einem der Sets lassen das andere unberührt.

```python
mein_set = {1, 2, 3}
mein_set2 = mein_set.copy()
```

```python
mein_set2
```

```python
mein_set
```

```python
mein_set.add(4)
```

```python
mein_set
```

```python
mein_set2
```

## Differenz

Um herauszufinden, was zwei Sets unterscheidet verwenden wir `.difference`. Die Syntax ist:

    set1.difference(set2)
    
Zum Beispiel:

```python
mein_set.difference(mein_set2)
```

## difference_update

Die Syntax von `. difference_update` ist:

    set1.difference_update(set2)
    
Diese Methode gibt `set1` zurück, nachdem die Elemente entfernt wurden, die in `set2` enthalten sind:

```python
mein_set = {1, 2, 3}
```

```python
mein_set2 = {1, 4, 5}
```

```python
mein_set.difference_update(mein_set2)
```

```python
mein_set
```

## Element verwerfen

Sofern ein Element Bestandteil eines Sets ist, kann es mit `.discard` entfernt werden. Andernfalls passiert nichts.

```python
mein_set
```

```python
mein_set.discard(2)
```

```python
mein_set
```

## Überschneidung

`.intersection` und `.intersection_update` geben die Überschneidung von zwei oder mehreren Sets als neues Set zurück.

```python
mein_set = {1, 2, 3}
```

```python
mein_set2 = {1, 2, 4}
```

```python
mein_set.intersection(mein_set2)
```

```python
mein_set
```

`.intersection_update` aktualisiert die Überschneidung in-place:

```python
mein_set.intersection_update(mein_set2)
```

```python
mein_set
```

## Keine Überschneidung

Um zu überprüfen, ob zwei Sets keine Schnittmenge haben, verwenden wir `.isdisjoint`. Diese Methode gibt `True`  aus, falls keine Schnittmenge vorliegt.

```python
mein_set = {1, 2}
mein_set2 = {1, 2, 4}
mein_set3 = {5}
```

```python
mein_set.isdisjoint(mein_set2)
```

```python
mein_set.isdisjoint(mein_set3)
```

## Teilmenge - Subset 
Die Methode `.issubset` ermittelt, ob ein anderes Set ein bestimmtes Set beinhaltet.

```python
mein_set
```

```python
mein_set2
```

```python
mein_set.issubset(mein_set2)
```

## Teilmenge - Superset

Die Methode `.issuperset` ermittelt, ob ein bestimmtes Set ein anderes Set beinhaltet.

```python
mein_set2.issuperset(mein_set)
```

```python
mein_set.issuperset(mein_set2)
```

 ## Symmetrie
 
`.symmetric_difference` und `.symmetric_update` geben den symmetrischen Unterschied als ein neues Set zurück, d.h. all die Elemente, die (nur) in genau einem der Sets beinhaltet sind.

```python
mein_set
```

```python
mein_set2
```

```python
mein_set.symmetric_difference(mein_set2)
```

## Vereinigung - Union

Die Methode `.union` vereinigt zwei Sets, d.h. bildet ein Set aller Werte, die in mindestens einem der beiden Sets auftauchen.

```python
mein_set.union(mein_set2)
```

## Update

Um ein Set mit der Vereinigung von sich und einem anderen Set zu überschreiben verwenden wir `.update`.

```python
mein_set.update(mein_set2)
```

```python
mein_set
```
