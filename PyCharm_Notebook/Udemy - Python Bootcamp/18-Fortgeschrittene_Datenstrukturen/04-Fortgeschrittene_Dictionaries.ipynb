# Fortgeschrittene Dictionaries

Im Gegensatz zu den anderen Datenstrukturen haben wir die meisten der wirklich nützlichen Methoden von Dictionaries während unseres Kurses bereits kennengelernt. Für die Vollständigkeit hier noch einige weitere:

## Dictionaryabstraktion

Genau wie bei der Listenabstraktion haben auch Dictionaries ihre eigene Version der Abstraktion (engl. comprehension), um sie schnell erstellen zu können. Es wird nicht so häufig verwendet, dennoch wollen wir ein Beispiel betrachten:

```python
{x: x**2 for x in range(10)}
```

Einer der Gründe dafür, dass es nicht so häufig verwendet wird ist, dass es schwer ist, Keys zu konstruieren, sofern sie nicht auf den Werten basieren.

## Iteration

Wir können über Dictionaries mittels der Methoden `.keys`, `.values` und `.items` iterieren. Zum Beispiel:

```python
dictionary = {'k1': 1, 'k2': 2}
```

```python
for key in dictionary.keys():
    print(key)
```

```python
for value in dictionary.values():
    print(value)
```

```python
for item in dictionary.items():
    print(item)
```

## Anzeige von Dictionary-Elementen

Die Methoden `.keys`, `.values` und `.items` geben ein sogenanntes *view object* zurück. Das ist keine unabhängige Liste der Elemente, sondern wird mit dem originalen Dictionary aktualisiert:

```python
key_view = dictionary.keys()

key_view
```

```python
dictionary['k3'] = 3

dictionary
```

```python
key_view
```
