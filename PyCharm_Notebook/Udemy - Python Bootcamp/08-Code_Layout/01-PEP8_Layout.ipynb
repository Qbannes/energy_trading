# Exkurs: Codeformatierung nach PEP8

Dir ist vielleicht schon aufgefallen, das wir in unseren Codebeispielen einer einheitlichen Formatierung folgen. Diese entsprechen dem gängigen Standard [PEP8](https://www.python.org/dev/peps/pep-0008/) (Python Enhancement Proposal 8, etwa Python-Ergänzungsvorschlag 8). Einheitliche Formatierung hilft es Code zu schreiben, der auch für andere Programmierer gut verständlich ist. Das ist besonders wichtig bei einer weit verbreiteten Programmiersprache wie Python, die in enorm vielen, kollaborativen Programmierprojekten eingesetzt wird. Ebenfalls ist in vielen Unternehmen,in denen mit Python programmiert wird, PEP8 die vorausgesetzte Konvention für die Codeformatierung, um die Zusammenarbeit zu erleichtern.

In diesem Notebook findest du eine kleine Einführung zum Thema PEP8. Wenn du die wenigen hier vorgestellten grundlegenden Formatierungsregeln beachtest, werden es dir andere Programmierer, die mit deinem Code arbeiten wollen, danken!

Die vollständige Übersicht zu allen Formatierungsregeln nach Pep8 findest du im Link zu Beginn dieser Textzelle.

## 1. Einrückung

Einrückung sollte immer mit vier Leerzeichen pro Codeblock erfolgen und niemals mit Tabulator.

```python
# SCHLECHT!
def test():
  while True:
    pass

# GUT!
def test():
    while True:
        pass
```

## 2. Zeichen pro Zeile

Eine einzelne Zeile Code sollte nie mehr als 80 Zeichen haben. Viele Editoren bieten es an, eine Markierung an dieser Grenze zu setzen, was dies erleichtert.

```python
# SCHLECHT!
satz = "You don’t frighten us, English pig dogs. Go and boil your bottoms, you sons of a silly person. I blow my nose at you, so-called 'Arthur King', you and all your silly English K-nig-hts."
# GUT!
satz = """You don’t frighten us, English pig dogs. Go and boil your bottoms, 
you sons of a silly person. I blow my nose at you, so-called 'Arthur King', 
you and all your silly English K-nig-hts.
""" 
```

## 3. Leerzeilen

Vor und nach Funktionen und Klassen der obersten Ebene sollten immer zwei Leerzeilen stehen. Möglichst selten kann eine einzelne Leerzeile zur logischen Trennung von Codeblöcken verwendet werden.

```python
# SCHLECHT!
a = 1
class Hallo:
    pass
def test():
    b = 2
    hallo = Hallo()
c = 3

# GUT!
a = 1


class Hallo:
    pass


def test():
    b = 2
    
    hallo = Hallo()
    
    
c = 3
```

## 5. Funktions- und Variablennamen

Bis auf wenige Ausnahmen (wie Schleifeniterationszähler `i, j` oder Funktionswerte `x, y` beispielsweise) sollten Variablennamen mit nur einem einzigen Zeichen vermieden werden. Verwende niemals `l` (kleines El), `O` (grosses Oh) oder `I` (grosses Ih) als Funktions- und Variablennamen.
Verwende sprechende Funktions- und Variablennamen, aber achte darauf, dass die Namen nicht zu lange werden. Weitere Sonderzeichen, auch deutsche Umlaute, sollten in Funktions- und Variablennamen nicht verwendet werden. Natürlich sollten ebenfalls keine reservierten Schlüsselwörter, wie zum Beispiel `str`, durch Variablen- oder Funktionsnamen ersetzt werden.

## 9. Docstrings

Für jedes Modul und jede Funktion, Klasse und Methode sollten Docstrings definiert werden. Docstrings in Python sind Strings, die eine Funktion oder Klasse erklären. Diese können mit `help(object)` oder `object.__doc__` ausgegeben werden, wobei `object` der Name der Klasse oder Funktion ist.<br>

Docstrings werden direkt in der Zeile unterhalb der Funktions- oder Klassendefinition geschrieben. Einzeilige Docstrings haben das folgende Format:<br>
`"""Text"""`<br>
Mehrzeilige Docstrings werden so formatiert:<br>
`""" Zeile 1
Zeile2 
"""`

```python
# SCHLECHT!
def test():
    pass

# SCHLECHT!
class Klasse:
    def test(self):
        pass

# GUT!
def test():
    """Funktion zum testen des Ablaufs."""
    pass

# GUT!
class Klasse:
    """Klasse zum testen des Programmes.
    Methode test() führt den eigentlichen Test aus.
    """
    
    def test(self):
        """Funktion zum testen des Ablaufs."""
        pass
```
