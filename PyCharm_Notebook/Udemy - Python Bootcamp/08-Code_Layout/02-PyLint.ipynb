# Automatisierte Codelayoutprüfung

Im vorherigen Notebook haben wir dir die gängige Konvention für das Layout von Python-Programmen, PEP8, vorgestellt. Es gibt einige automatisierte Tools, die deinen Code auf PEP8-Konformität prüfen können und teilweise auch kleinere Programmierfehler finden.

Erinnere dich daran, dass wir die Inhalte einer Zelle mit der IPython-Magic `%%writefile` in eine Datei schreiben können. Weiterhin können wir Terminalbefehle in einer Zelle mit einem vorangestellten `!` ausführen.


## Layout-Werkzeuge

Die Werkzeuge zur Überprüfung deiner Codeoformatierung stammen von Fremdanbietern und müssen daher zuerst installiert werden, wie beispielsweise:

* [pylint](https://www.pylint.org/)
* [pyflakes](https://pypi.python.org/pypi/pyflakes/)
* [pep8](https://pypi.python.org/pypi/pep8)

Das sind einfache Werkzeuge, die lediglich deinen Code analysieren und dir mitteilen, ob es Probleme mit der Formatierung gibt oder anderen Kleinigkeiten, wie zum Beispiel der Zugriff auf Variablen vor deren Definition.

Wir beschäftigen uns in diesem Notebook mit `PyLint`.

```python
%%writefile einfach1.py
"""
Ein sehr einfaches Script.
"""

def meine_func():
    """ Eine Extrem einfache Funktion. """
    
    erste = 1
    zweite = 2
    
    print(erste)
    print(zweite)

meine_func()
```

Viel besser! Unser Ergebnis ist 6.67 von 10. Die verbleibende Formatierungsschwäche sind überzählige Leerzeichen, das ist nicht weiter schlimm. `pylint` hat uns geholfen, unser Programm wesentlich zu verbessern. Aber was ist mit komplexeren Problemen?
