**AUFGABE: Dir ist vielleicht aufgefallen, dass es mehr als eine Seite gibt und das Folgeseiten aussehen wie http://quotes.toscrape.com/page/2/. Wende dein Wissen über for-Schleifen und Stringkonkatenation an, um über alle Seiten zu iterieren und alle Autorennamen zu extrahieren. Es gibt viele Wege, dies zu erreichen, aber deine Schleife wird erkennen müssen, wenn die letzte Seite erreicht ist. Gehe für das Debugging davon aus, dass die Webseite nur 10 Seiten hat, also ist die letzte Seite http://quotes.toscrape.com/page/10/. Versuche aber, eine robuste Schleife zu implementieren, die bei einer unbekannten Seitenzahl auch funktionieren würde. Ein möglicher Ansatz ist try/except.**

### Lösungsmöglichkeit 1 ( bei bekannter Seitenzahl)

```python
url = 'http://quotes.toscrape.com/page/'
```

```python
autoren = set()

for seite in range(1, 10):

    # Konkatenation für neue Seiten-URL
    seiten_url = url + str(seite)
    # Response erzeugen
    res = requests.get(seiten_url)
    # soup erzeugen
    soup = bs4.BeautifulSoup(res.text, 'lxml')
    
    # Autoren zum Set hinzufügen
    for name in soup.select(".author"):
        autoren.add(name.text)
```

```python
autoren
```

### Lösungsmöglichkeit 2 (Unbekannte Seitenzahl)

Wie sieht eine ungültige Seitennummer aus?

```python
# Wähle eine enorme Seitenzahl, die sicher nicht existiert
page_url = url + str(9999999)

res = requests.get(page_url)

soup = bs4.BeautifulSoup(res.text, 'lxml')
```

```python
soup
```

```python
# Diese Lösung setzt voraus, dass der String "No quotes found!" nur auf der letzten
# (ungültigen) Seite vorkommt. Falls dieser String auf einer früheren Seite auch
# vorkommt, müssen wir eine robustere Methode implementieren.
"No quotes found!" in res.text
```

```python
seite_gueltig = True
autoren = set()
seite = 1

while seite_gueltig:

    # Konkatenation für neue Seiten-URL
    seiten_url = url + str(seite)

    res = requests.get(seiten_url)

    # Prüfe, ob wir uns auf der letzten (ungültigen) Seite befinden
    if "No quotes found!" in res.text:
        break

    soup = bs4.BeautifulSoup(res.text, 'lxml')

    # Autoren zum Set hinzufügen
    for name in soup.select(".author"):
        autoren.add(name.text)

    # Gehe zur nächsten Seite
    seite += 1
```

```python
autoren
```

Es gibt viele  Lösungsmöglichkeiten, viele davon auch robuster und flexibler. Der Grundgedanke ist aber derselbe: Iteration mit einer Schleife über alle möglichen Seiten und Abbruch beim erreichen einer ungültigen Seite.
