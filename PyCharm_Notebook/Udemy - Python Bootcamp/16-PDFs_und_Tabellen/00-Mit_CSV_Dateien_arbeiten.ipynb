# Mit CSV-Dateien arbeiten

Willkommen zurück! Lass uns herausfinden, wie man in Python mit CSV-Dateien arbeitet. CSV steht für Comma Seperated Values, per Komma getrennte Werte. Alle CSV-Dateien sind einfache Textdateien mit alphanumerischen Zeichen. Der Unterschied von CSV-Dateien zu den ähnlich strukturierten Excel-Dateien ist, dass CSV keine Datentypen für die Werte unterstützt - alle Werte sind Strings. Es gibt auch keine Arbeitsblätter wie bei Excel. Python bietet einige Bibliotheken für die Arbeit mit Excel-Dateien, siehe [hier](http://www.python-excel.org/) und [hier](https://www.xlwings.org/).

CSV-Dateien werden gerne für den Transfer grosser Datenmengen genutzt. Datenbanken, Analyseprogramme und andere Programme, die mit massiven Datenmengen arbeiten (wie Kontakte und Kundendaten) unterstützen normalerweise den Ex- und Import in CSV.

Wir schauen jetzt, wie wir eine CSV-Datei mit der entsprechenden Python-Standardbibliothek öffnen können.

### Encoding

CSV-Dateien enthalten häufig Zeichen, die von Python nicht verstanden werden, z.B. Sonderzeichen von anderen Sprachen. Wirf hierzu einen Blick auf diese Diskussion auf [Stackoverflow](https://stackoverflow.com/questions/9233027/unicodedecodeerror-charmap-codec-cant-decode-byte-x-in-position-y-character).

Standardmäßig ist das Encoding in **UTF-8**. Bekommst du Errors mit einem Hinweis **can't decode**, dann könnte das daran liegen, dass die CSV-Datei ein anderes Encoding benötigt.

```python
csv_data = csv.reader(data)
data_zeilen = list(csv_data)
```

```python
data_zeilen[:3]
```

Wie du siehst enthält die erste Zeile die Spaltenüberschriften. Wir können das besser formatieren:

```python
for zeile in data_zeilen[:5]:
    print(zeile)
```

Wie können wir daraus beispielsweise alle Emailadressen extrahieren? Zur Demonstration machen wir das nur für ein paar Zeilen, da der Datensatz über 1000 Zeilen enthält.

```python
len(data_zeilen)
```

```python
alle_emails = []

for zeile in data_zeilen[1:15]:
    alle_emails.append(zeile[3])
```

```python
print(alle_emails)
```

Und wenn wir statt dessen eine Liste der vollständigen Namen haben wollen?

```python
namen = []

for zeile in data_zeilen[1:15]:
    namen.append(zeile[1] + ' ' + zeile[2])
```

```python
namen
```

## In CSV-Dateien schreiben

Wir können auch CSV-Dateien neu erzeugen oder vorhandene ergänzen.

### Neue Datei
**Dies überschreibt bestehende Dateien, sei also vorsichtig in der Anwendung!**

```python
# newline kontrolliert, wie Zeilen umgebrochen werden.
# Möglich sind None, '', '\n', '\r', and '\r\n'
ausgabedatei = open('csv_dateiname.csv', 'w', newline='')
```

```python
csv_writer = csv.writer(ausgabedatei, delimiter=',')
```

```python
csv_writer.writerow(['a', 'b', 'c'])
```

```python
csv_writer.writerows([['1', '2', '3'], ['4', '5', '6']])
```

```python
ausgabedatei.close()
```

____
### Existierende Datei

```python
f = open('csv_dateiname.csv', 'a', newline='')
```

```python
csv_writer = csv.writer(f)
```

```python
csv_writer.writerow(['neu', 'neu', 'neu'])
```

```python
f.close()
```
