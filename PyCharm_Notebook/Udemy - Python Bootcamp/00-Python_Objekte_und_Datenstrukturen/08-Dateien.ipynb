## Datei öffnen mit Python

Lasst uns die Datei test.txt öffnen, die im selben Verzeichnis wie dieses Notebook liegt. Im folgenden werden wir nur mit Dateien im lokalen Verzeichnis arbeiten.

Es ist leicht, in diesem Schritt einen Fehler zu erzeugen:

```python
datei = open('whoops.txt')
```

Stelle zur Vermeidung dieses Fehlers sicher, dass die zu öffnende .txt-Datei im selben Verzeichnis wie das Notebook liegt. Das kannst du mit **pwd** überprüfen:

**Alternativ kannst du den gesamten Dateipfad übergeben, um auf beliebige Verzeichnisse und Dateien auf deinem Computer zuzugreifen.**

Windows benötigt dabei doppelte \, damit Python den zweiten \ nicht als Maskierungszeichen versteht, also

    datei = open("C:\\Beliebiger\\Pfad\\datei.txt")
    
MacOS und Linux verwendet statt \ den /:

    datei = open("/Beliebiger/Pfad/datei.txt")

```python
# Die Datei test.txt öffnen, die wir gerade erstellt haben
datei = open('test.txt')
```

```python
# Jetzt können wir den Inhalt der Datei lesen
datei.read()
```

```python
# Was passiert aber, wenn wir erneut lesen wollen?
datei.read()
```

Das passiert, weil man sich Python beim Lesen als Cursor vorstellen kann, der bis zum Ende der Datei geht. Dort bleibt er stehen und befindet sich dort nach dem ersten read Befehl. Es gibt also nichts mehr zu lesen. Wir können den Cursor folgendermaßen zurücksetzen:

```python
# Den Anfang der Datei (Index 0) suchen
datei.seek(0)
```

```python
# Jetzt kann die Datei erneut gelesen werden
datei.read()
```

Um nicht immer wieder zum Anfang zu müssen können wir auch die `readlines` Funktion nutzen. Bei großen Dateien sollte man allerdings Vorsicht walten lassen, da alles im Speicher gehalten wird. Wie man mit großen Dateien umgehen kann schauen wir uns später im Kurs an.

```python
# readlines() gibt eine Liste der Zeilen der Datei zurück
datei.seek(0)
datei.readlines()
```

Nach der Verwendung einer Datei sollte sie wieder geschlossen werden:

```python
datei.close()
```

## In eine Datei schreiben

Die Voreinstellung der Funktion `open` erlaubt es nur, Dateien zu lesen. Wir müssen den Parameter 'w' übergeben, um Inhalte überschreiben zu können. Zum Beispiel:

```python
# Wir fügen einen zweiten Parameter 'w' hinzu, der für Schreiben (write) steht
# 'w+' anstelle dessen gestattet Lese- und Schreibzugriff
datei = open('test.txt', 'w')
```

### <strong><font color='red'>Vorsicht!</font></strong> 
Das Öffnen einer Datei mit `'w'` or `'w+'` leert das Original, der vorherige Inhalt der Datei **wird gelöscht**!

```python
# In die Datei schreiben
datei.write('Das ist eine neue Zeile')
```

```python
# Die Datei lesen
datei.seek(0)
datei.read()
```

```python
datei.close()  # Führe dies immer aus, wenn du fertig bist mit der Datei
```

## Eine Datei ergänzen
Der Parameter `a` öffnet die Datei und setzt den Schreibkopf ans Ende, so dass alles, was in die Datei geschrieben wird, am Ende angehängt wird. Wie `w+` erlaubt `a+` das Lesen und Schreiben einer Datei. Wenn die Datei nicht existiert, wird sie erzeugt.

```python
datei = open('test.txt', 'a+')
datei.write('\nDieser Text wird test.txt angehängt')
datei.write('\nUnd eine weitere Zeile hier')
```

```python
datei.seek(0)
print(datei.read())
```

```python
datei.close()
```

### Ergänzen mit `%%writefile`
Wir können das selbe mit IPython-Magic ausführen:

```python
%%writefile -a test.txt

Dieser Text wird test.txt angehängt
Und eine weitere Zeile hier
```

Verwende ein Leerzeichen, wenn die erste Zeile in einer eigenen Zeile beginnen soll, da Jupyter maskierte Zeichen wie `\n` nicht kennt.

## Iterationen in einer Datei

Es folgte eine kurze Vorschau einer for-Schleife, mit der wir Iterationen in einer Datei ausführen. Zunächst erstellen wir mit iPython eine neue Datei:

```python
%%writefile test.txt
Erste Zeile
Zweite Zeile
```

Jetzt können wir unserem Programm sagen, durch jede Zeile der Datei zu gehen und etwas zu tun:

```python
for line in open('test.txt'):
    print(line)
```

Macht euch keine Sorgen darüber, hier bereits alles komplett zu verstehen. For-Schleifen werden bald behandelt. Aber wir können schon einmal beleuchten, was wir oben gemacht haben. Wir haben programmiert für jede Zeile der Datei die Zeile auszugeben. Wichtige Punkte hierbei sind:

1. Wir hätten 'zeile' jegliche Bezeichnungen geben können (siehe nächstes Beispiel).
2. Da wir nicht .read() ausgeführt haben, wurde die Datei nicht in den Speicher geladen.
3. Achtet auf die Einrückung der zweiten Zeile Code. Das ist in Python notwendig.

```python
# Siehe obigen Punkt 1.
for asdf in open('test.txt'):
    print(asdf)
```
