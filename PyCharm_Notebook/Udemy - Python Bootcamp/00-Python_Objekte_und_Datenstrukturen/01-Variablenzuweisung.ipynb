# Variablenzuweisung

## Regeln für Variablennamen
* Namen dürfen nicht mit einer Nummer starten
* Namen dürfen keine Leerzeichen enthalten, verwende statt dessen _
* Namen dürfen keins dieser Zeichen enthalten:

      :'",<>/?|\!@#%^&*~-+

* Best Practise ([PEP8](https://www.python.org/dev/peps/pep-0008/#function-and-variable-names)): Variablennamen in Kleinschreibung und mit Unterstrichen
* vermeide eingebaute Python-Keywords wie `list` und `str`
* vermeide die Verwendung der Zeichen `l` (kleines El), `O` (großes Oh) and `I` (großes Ie), da diese mit `1` und `0` verwechselt werden können

## Dynamic Typing

Python verwendet *dynamic typing* (dynamische Typisierung), was bedeutet, dass du einer einzigen Variablen verschiedene Datentypen zuweisen kannst. Dadurch ist Python sehr flexibel in der Handhabung von Datentypen, im Gegensatz zu anderen Sprachen mit *static typing* (statischer Typisierung). 

```python
meine_hunde = 2
```

```python
meine_hunde
```

```python
meine_hunde = ['Sammy', 'Frankie']
```

```python
meine_hunde
```

### Vor- und Nachteile von Dynamic Typing
#### Vorteile
* einfach in der Anwendung
* schnellere Entwicklung

#### Nachteile
* kann unerwartete Bugs hervorrufen!
* muss gelegentlich mit `type()` untersucht werden.

## Variablenzuweisung
Variablenzuweisung hat die Syntax `name = Objekt`, mit einem einzelnen Gleichheitszeichen `=` als *Zuweisungsoperator*.

```python
a = 5
```

```python
a
```

Hier haben wir der Variable `a` das Integerobjekt `5` zugewiesen.
Lasst uns `a` etwas anderes zuweisen:

```python
a = 10
```

```python
a
```

Du kannst jetzt `a` anstelle der Zahl `10` verwenden:

```python
a + a
```

## Wiederholte Variablenzuweisung
Python erlaubt dir die wiederholte Zuweisung von Variablen mit einer Referenz auf das selbe Objekt.

```python
a = a + 10
```

```python
a
```

Dafür gibt es eine Abkürzung. Python erlaubt die Addition, Subtraktion, Multiplikation und Divison bei gleichzeitiger Zuweisung mit `+=`, `-=`, `*=`, und `/=`.

```python
a += 10
```

```python
a
```

```python
a *= 2
```

```python
a
```

## Variablentyp mit `type()` bestimmen
Mit der In Python eingebauten Funktion `type()` kannst du den Typ eines Objektes überprüfen. Gebräuchliche Datentypen sind unter anderem
* **int** (für Integer)
* **float**
* **str** (für String)
* **list**
* **tuple**
* **dict** (für Dictionary)
* **set**
* **bool** (für Bool'sches True/False)

```python
type(a)
```

```python
a = (1, 2)
```

```python
type(a)
```
