## Arten von Zahlen

Python beinhaltet verschiedene "Arten" von Zahlen. Wir werden uns hauptsächlich mit Ganzzahlen (en.: integers) und Gleitkommazahlen (en.: floating point numbers) beschäftigen. Später unternehmen wir einen kurzen Abstecher zu den komplexen Zahlen (en.: complex numbers).
Integers sind einfach ganze Zahlen, positiv und negativ. Zum Beispiel: 2 und -2 sind Integers.
Floats sind Kommazahlen und in Python besonders, da sie einen Dezimalpunkt oder einen Exponenten (e) beinhalten, um die Zahl zu definieren. Zum Beispiel sind 2.0 und -2.1 Zahlen mit Dezimalpunkt. 4E2 (4 mal 10 hoch 2) ist ebenso ein Beispiel für eine Float-Zahl in Python.

Im Laufe dieses Kurses werden wir insbesondere mit Integern und Floats arbeiten.

## Grundlegende Arithmetik

```python
# Addition
2 + 1
```

```python
# Subtraktion
2 - 1
```

```python
# Multiplikation
2 * 2
```

```python
# Division
3 / 2
```

```python
# Integerdivision
7 // 4
```

**Whoa! Was ist hier passiert? 7 geteilt durch 4 ist doch 1.75 und nicht 1!**

Das liegt daran, das wir eine "*Integerdivision*" durchführen. Der Operator // schneidet die Dezimalstellen ohne Rundung ab und gibt ein Integer als Ergebnis zurück.

**Und wenn wir nur den Rest der Division wissen wollen?**

```python
# Modulo
7 % 4
```

4 passt genau einmal in 7 mit einem Rest von 3. Der Modulo-Operator % gibt diesen Rest zurück.

### Weitere Arithmetik

```python
# Potenz
2 ** 3
```

```python
# So kann ebenfalls die Wurzel gezogen werden
4 ** 0.5
```

```python
# Die Reihenfolge der Operatoren wird in Python eingehalten
2 + 10 * 10 + 3
```

```python
# Klammern können genutzt werden um die Reihenfolge zu spezifizieren
(2 + 10) * (10 + 3)
```
