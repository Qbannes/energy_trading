# Stringformatierung

Stringformatierung erlaubt es uns, Elemente in einen String einzufügen, anstatt diese mit Komma und Plus zusammenzufügen. Hier ein schneller Vergleich:


    spieler = 'Thomas'
    punkte = 33
    
    'Letzte Nacht hat ' + spieler + ' insgesamt ' + str(punkte) + ' Punkte gesammelt.'  # Konkatenation
    
    f'Letzte Nacht hat {player} insgesamt {punkte} Punkte gesammelt.'          # Stringformatierung


Es gibt drei Methoden für die Stringformatierung.
* Die älteste Methode verwendet Platzhalter mit dem Modulo-Zeichen `%`
* Angenehmer in der Anwendung ist die String-Methode `.format()`.
* Die neuste, mit Python 3.6 eingeführte Methode, verwendet *f-Strings* genannte, formatierte Stringliterale.

Da du wahrscheinlich bei der Arbeit mit Code von Anderen auf alle drei Methoden treffen wirst, beschreiben wir hier alle drei. 

## Formatierung mit Platzhaltern
Du kannst <code>%s</code> verwenden, um Strings in dein Print-Statement einzufügen. Der Modulo-Operator `%` wird in diesem Zusammenhang als "Stringformatierungsoperator" bezeichnet.

```python
print("Hier werde ich %s einfügen." % 'etwas')
```

Mehrere Elemente können mit einem Tupel nach dem `%`-Operator eingefügt werden.

```python
print("Ich werde hier %s Text einfügen und einen %s hier." % ('einen','weiteren'))
```

Du kannst auch Variablennamen übergeben:

```python
x, y = 'einen', 'weiteren'
print("Ich werde hier %s Text einfügen und einen %s hier." % (x, y))
```

### Formatkonvertierungsmethoden
An dieser Stelle ist anzumerken, dass die Operatoren <code>%s</code> und <code>%r</code> ein beliebiges Objekt in einen String konvertieren anhand der Methoden `str()` und `repr()`. Wir werden in späteren Kapiteln mehr über diese beiden Funktionen lernen, du solltest aber beachten, dass `%r` und `repr()` die *Stringrepräsentation* des Objektes zurückgeben, inklusive Hochkommata und Maskierungszeichen.

```python
print('Er sagte, sein Name sei %s.' % 'Fred')
print('Er sagte, sein Name sei %r.' % 'Fred')
```

Mit dem maskierten t `\t` wird ein Tabulator eingefügt: 

```python
print('Wir sind %s.' % 'am \tbraten')
print('Wir sind %r.' % 'am \tbraten')
```

Der Operator `%s` konvertiert alles in einen String, inklusive Integers und Floats. Der Operator `%d` konvertiert zuerst Zahlen in Integer, allerdings ohne dabei zu runden. Siehe hier:

```python
print('Ich schrieb %s Programme heute.' % 3.75)
print('Ich schrieb %d Programme heute.' % 3.75)
```

### Einrückung und Präzision von Gleitkommazahlen

Gleitkommazahlen (Floats) verwenden das Format <code>%5.2f</code>. Hier entspricht <code>5</code> der minimalen Anzahl von Zeichen, die der String enthalten soll; diese werden mit Leerzeichen aufgefüllt, wenn die gesamte Zahl zu wenige Stellen hat. Das folgende <code>.2f</code> steht für die Anzahl der Nachkommastellen. Hier sind ein paar Beispiele:

```python
print('Gleitkommazahl: %5.2f' % (13.144))
```

```python
print('Gleitkommazahl: %1.0f' % (13.144))
```

```python
print('Gleitkommazahl: %1.5f' % (13.144))
```

```python
print('Gleitkommazahl: %10.2f' % (13.144))
```

```python
print('Gleitkommazahl: %25.2f' % (13.144))
```

Mehr Informationen zur Stringformatierung mit Gleitkommazahlen findest du auf https://docs.python.org/3/library/stdtypes.html#old-string-formatting

### Mehrfache Formatierung
Nichts verbietet es, mehr als einen Formatierungsoperator einzusetzen:

```python
print('Erster: %s, Zweiter: %5.2f, Dritter: %r' % ('Hi!', 3.1415, 'Ciao!'))
```

## Formatierung mit der Methode `.format()`
Ein besser Weg zum Einfügen von Elementen in Print-Statements ist die Stringmethode `.format()`. Die Syntax ist:

    'String hier {} dann auch {}'.format('etwas1','etwas2')
    
Zum Beispiel:

```python
print('Dies ist ein String mit einer {}'.format('Einfügung'))
```

### Die Methode .format() hat einige Vorteile gegenüber dem Operator %s:

#### 1. Die Einfügung kann per Indizierung erfolgen:

```python
print('Der {2} {1} {0}'.format('Fuchs', 'braune', 'schnelle'))
```

#### 2. Es kann auch mittels Variablen eingefügt werden:

```python
print('Erstes Objekt: {a}, Zweites Objekt: {b}, Drittes Objekt: {c}'.format(
    a=1, b='Zwei', c=12.3))
```

#### 3. Elemente können wiederholt verwendet werden:

```python
print('Ein gesparter %s ist ein verdienter %s.' % ('Cent', 'Cent'))
# vs.
print('Ein gesparter {c} ist ein verdienter {c}.'.format(c='Cent'))
```

### Ausrichtung, Auffüllen und Präzision mit `.format()`
Mit geschweiften Klammern können Feldlängen, Ausrichtung nach Links/Rechts, Rundungsparameter und mehr festgelegt werden.

```python
print('{0:8} | {1:9}'.format('Frucht', 'Anzahl'))
print('{0:8} | {1:9}'.format('Äpfel', 3.))
print('{0:8} | {1:9}'.format('Orangen', 10))
```

Die Voreinstellung von `.format()` ist die Ausrichtung von Text nach links und Zahlen nach rechts. Mit den optionalen Parametern `<`,`^`, oder `>` kann die Ausrichtung nach links, mittig oder rechts definiert werden:

```python
print('{0:<8} | {1:^8} | {2:>8}'.format('Links', 'Mittig', 'Rechts'))
print('{0:<8} | {1:^8} | {2:>8}'.format(11, 22, 33))
```

Du kannst den Ausrichtungsoperator mit einem Auffülloperator erweitern:

```python
print('{0:=<8} | {1:-^8} | {2:.>8}'.format('Links', 'Mittig', 'Rechts'))
print('{0:=<8} | {1:-^8} | {2:.>8}'.format(11, 22, 33))
```

Feldlänge und Nachkommazahlen werden ähnlich wie bei der ersten Methode gehandhabt. Die folgenden beiden Print-Statements sind äquivalent:

```python
print('Dies ist meine zweistellige Gleitkommazahl mit zehn Zeichen: %10.2f' % 13.579)
print(
    'Dies ist meine zweistellige Gleitkommazahl mit zehn Zeichen: {0:10.2f}'.format(13.579))
```

Beachte: Auf den Doppelpunkt folgen 5 Leerzeichen und 5 Zeichen für 13.58 insgesamt also 10 Zeichen. 

Mehr Informationen zur Stringmethode `.format()` findest du auf https://docs.python.org/3/library/string.html#formatstrings

## Formatierte Stringliterale (f-strings)

Die in Python 3.6 eingeführten f-Strings haben verschiedene Vorteile gegenüber den älteren, obig beschriebenen Methoden der Stringformatierung. Beispielsweise kannst du normale Variablen direkt im String einbinden, ohne sie als Parameter vorher mit `.format(var)` zu übergeben.

```python
name = 'Fred'

print(f"Er sagte, sein Name sei {name}.")
```

Übergib `!r` für die String-Repräsentation:

```python
print(f"Er sagte, sein Name sei {name!r}")
```

#### Gleitkommazahlformatierung mittels `"result: {value:{width}.{precision}}"`

Der Ausdruck `{value:10.4f}` der Methode `.format()` entspricht dem f-String `{value:{10}.{6}}`:

```python
num = 23.45678
print(
    "Dies ist meine vierstellige Gleitkommazahl mit zehn Zeichen: {0:10.4f}".format(num))
print(
    f"Dies ist meine vierstellige Gleitkommazahl mit zehn Zeichen: {num:{10}.{6}}")
```

Beachte, dass sich *Präzision* bei f-Strings auf die Gesamtzahl der Stellen der Zahl bezieht, nicht nur auf die Nachkommastellen!
Das passt besser zur wissenschaftlichen Zahlennotation und statistischer Analyse. Unglücklicherweise unterstützen f-Strings nicht das Auffüllen von rechts, selbst wenn Präzision es erlaubt:

```python
num = 23.45
print(
    "Dies ist meine vierstellige Gleitkommazahl mit zehn Zeichen: {0:10.4f}".format(num))
print(
    f"Dies ist meine vierstellige Gleitkommazahl mit zehn Zeichen: {num:{10}.{6}}")
```

Wenn es erforderlich ist, kannst du immer die `.format()`-Snytax innerhalb eines f-Strings anwenden:

```python
num = 23.45
print(
    "Dies ist meine vierstellige Gleitkommazahl mit zehn Zeichen: {0:10.4f}".format(num))
print(
    f"Dies ist meine vierstellige Gleitkommazahl mit zehn Zeichen: {num:10.4f}")
```
