## In Email suchen

Nach der erfolgreichen Verbindung mit dem Mailserver können wir mit der spezialisierten Syntax von IMAP in unseren Mails suchen. Hier sind die wichtigsten Schlüsselwörter für die Suche:

<table border='1'>
    <tr>
        <th align='center'>Schlüsselwort</th>
        <th align='center'>Definition</th>
    </tr>
    <tr>
        <td>'ALL'</td>
        <td>
            Gibt alle Nachrichten in deinem Emailverzeichnis zurück. Gegebenenfalls ist das Größenlimit von
            imaplib zu beachten. Dieses kann angepasst werden mit imaplib._MAXLINE = 100, wobei du statt 100
            ein beliebiges Limit setzen kannst.
        </td>
    </tr>
    <tr>
        <td>'BEFORE datum'</td>
        <td>
            Gibt alle Nachrichten vor datum zurück. datum muss im Format 1-Nov-2000 vorliegen.
        </td>
    </tr>    
    <tr>
        <td>'ON datum'</td>
        <td>
        Gibt alle Nachrichten am Tag datum zurück. datum muss im Format 1-Nov-2000 vorliegen.
        </td>
    </tr>
     <tr>
        <td>'SINCE datum'</td>
        <td>
        Gibt alle Nachrichten bis datum zurück. datum muss im Format 1-Nov-2000 vorliegen.
        </td>
    </tr>
    <tr>
        <td>'FROM absender '</td>
        <td>
        Gibt alle Nachrichten des Absenders absender zurück. Kann eine vollständige Emailadresse
        oder nur ein Teil der Adresse sein, also "nutzer@test.com" oder nur z.B. nur "test".
        </td>
    </tr>
    <tr>
        <td>'TO empfaenger'</td>
        <td>
        Gibt alle an empfaenger versendete Nachrichten zurück. Kann eine vollständige Emailadresse
        oder nur ein Teil der Adresse sein, also "nutzer@test.com" oder nur z.B. nur "test".
        </td>
    </tr>
    <tr>
        <td>'CC name' und/oder 'BCC name'</td>
        <td>
        Gibt alle Emails mit name als CC und/oder name als BCC zurück. Kann eine vollständige Emailadresse
        oder nur ein Teil der Adresse sein, also "nutzer@test.com" oder nur z.B. nur "test".
        </td>
    </tr> 
    <tr>
        <td>'SUBJECT string','BODY string','TEXT "string mit leerzeichen"'</td>
        <td>
        Gibt alle Nachrichten mit passendem string im Betreff oder im Body der Email zurück. 
        Strings mit Leerzeichen müssen in doppelte Anführungszeichen gepackt werden.
        </td>
    </tr>
    <tr>
        <td>'SEEN', 'UNSEEN'</td>
        <td>
            Gibt alle angesehenen oder nicht angesehenen Emails zurück (resp. gelesen/ungelesen).
        </td>
    </tr>
    <tr>
        <td>'ANSWERED', 'UNANSWERED'</td>
        <td>
            Gibt alle beantworteten oder unbeantworteten Emails zurück.
        </td>
    </tr>
    

</table>
