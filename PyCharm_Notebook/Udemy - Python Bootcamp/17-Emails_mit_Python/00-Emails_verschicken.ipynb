# Emails verschicken

```python
import smtplib
```

Jetzt erzeugen wir das SMTP-Objekt, welches Methoden zum Login bei deinem Emailprovider und anschliessendem Versand von Emails bereitstellt. Beachte, dass du eine Portnummer angeben musst. Falls der Port 587 nicht funktioniert, kannst du statt dessen die Portnummer 465 ausprobieren. Achtung: eine Firewall oder ein Antivirusprogramm auf deinem Computer könnte verhindern, dass du den Port verwenden kannst, weshalb du diese womöglich deaktivieren musst.

```python
smtp_object = smtplib.SMTP('smtp.gmail.com', 587)
```

Als nächstes müssen wir die Methode `.ehlo` ausführen, welches den Server ruft und die Verbindung einrichtet. Diese Methode soltle direkt nach dem erzeugen des SMTP-Objekts ausgeführt werden. Wenn du vorher andere Methoden aufrufst, kann das später zu Verbindungsproblemen führen. Das erste Element im zurückgegebenen Tupel sollte 250 sein, was bedeutet, dass die Verbindung erfolgreich war.

```python
smtp_object.ehlo()
```

Wenn du den Port 587 verwendest, läuft im Hintergrund eine TLS-Verschlüsselung. Diese muss mit der Methode `.starttls` initialisiert werden. Wenn du Port 465 benutzt, dann basiert die Verschlüsserlung auf SSL und dieser Schritt ist nicht auszuführen.

```python
smtp_object.starttls()
```

Jetzt konfigurieren wir den Emailaccount mit dem Passwort. Du solltest niemals dein Emailpasswort als einfachen String in einem Script speichern, da dann jeder mit Zugriff auf das Script dein Passwort kennt! Es ist also besser, das Passwort per Input zu erfragen, und am allerbesten verbergen wir noch die Eingabe mit der Standardbibliothek `getpass`. `getpass` zeigt dann die Eingabe versteckt hinter Punkten an, so wie du es von Webseiten-Passworteingaben kennst.

```python
# Für versteckte Passwörter
import getpass
```

```python
ergebnis = getpass.getpass("Gib hier etwas ein und es wird versteckt sein: ")
```

```python
# Aber beachte: Die Eingabe ist intern normal in einem String gespeichert!
ergebnis
```

```python
# Oder nimm einfach ein input()
input("Gib dein Passwort ein: ")
```

____
**Wenn du Gmail benutzen möchtest, musst du ein App-Passwort erzeugen, dein normales Passwort wird nicht funktionieren. Dabei ist eine 2-Schritt-Authentifikation notwenig. Die dafür benötigten Schritte werden hier erklärt: https://support.google.com/accounts/answer/185833?hl=en/. Richte die 2-Schritt-Authentifizierung ein, erzeuge das App-Passwort, wähle Mail als App und gib dem ganzen einen beliebigen Namen. Du erhältst dann ein Passwort mit 16 Zeichen. Verwende dieses Passwort als Login-Passwort für SMTP.**
____

```python
email = getpass.getpass("Gib deine Emailadresse ein: ")
passwort = getpass.getpass("Gib dein Passwort ein: ")
smtp_object.login(email, passwort)
```

Jetzt kannst du mit der Methode `.sendmail` eine Email verschicken.

```python
adresse_absender = getpass.getpass("Gib deine Emailadresse ein: ")
adresse_empfaenger = getpass.getpass(
    "Gib die Emailadresse des Empfängerns ein: ")
betreff = input("Gib den Betreff ein: ")
nachricht = input("Gib den Text deiner Email ein: ")
komplett = "Betreff: " + betreff + '\n' + nachricht
smtp_object.sendmail(adresse_absender, adresse_empfaenger, komplett)
```

Wenn du ein leeres Dictionary als Rückgabe erhältst, dann war der Versand erfolgreich.

Anschliessend kannst du die Verbindung beenden mit der Methode `.quit`.

```python
smtp_object.quit()
```

Nachdem wir jetzt wissen, wie eine Email versendet werden kann, ist es jetzt an der Zeit, einen Blick auf die von dir empfangenen Emails zu werfen.
