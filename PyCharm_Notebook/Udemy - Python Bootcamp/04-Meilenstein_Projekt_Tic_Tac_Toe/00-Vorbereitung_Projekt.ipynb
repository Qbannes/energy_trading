**Erweitere die Funktion um die Bestätigung der Gültigkeit der Eingabe**

```python
def nutzer_eingabe():
    """ Funktion, die vom Nutzer die Eingabe einer Zahl zwischen 1 und 10 erwartet
    und diese als Integerwert zurückgibt.
    """
    
    # Der Wert der Eingabe wird als falscher Wert initialisiert
    eingabe = 'falsch'
    
    # Frage nach erneuter Eingabe, solange die Zahl keiner Ziffer ist
    while eingabe.isdigit() == False:
        
        # Hier sollte nicht konvertiert werden, da wir sonst einen Fehler bei falschen Eingaben erzeugen
        eingabe = input("Bitte gib eine Zahl zwischen 1 und 10 ein: ")
    
    # Hier konvertieren wir, nachdem die while-Schleife erfolgreich beendet wurde
    return int(eingabe)
```

```python
def nutzer_eingabe():
    """ Funktion, die vom Nutzer die Eingabe einer Zahl zwischen 1 und 10 erwartet
    und diese als Integerwert zurückgibt.
    """
    
    eingabe = 'falsch'
    im_bereich = False
    
    while eingabe.isdigit() == False or im_bereich == False:
        
    
    
        eingabe = input("Bitte gib eine Zahl zwischen 1 und 10 ein: ")
        
        if eingabe.isdigit() == False:
            print("Leider hast du keine Zahl eingegeben!")
            
        if eingabe.isdigit() == True:
            if int(eingabe) in range(0, 10):
                im_bereich = True
            else:
                im_bereich = False
        
    
    return int(eingabe)
```

## Einfache Nutzerinteraktionen

**Lasst uns zum Schluss alle vorgestellten Konzepte kombinieren, um einen kleines Spiel zu programmieren, in dem der Benutzer eine "Position" in einer existierenden Liste wählt und mit einem Wert seiner Wahl ersetzt.**

```python
spiel_liste = [0, 1, 2]
```

```python
def liste_ausgeben(spiel_liste):
    """ Gibt die Eingabeliste aus. """
    
    print("Hier ist die derzeitige Liste")
    print(spiel_liste)
```

```python
def waehle_position():
    """ Funktion, die vom Nutzer die Eingabe einer Zahl 0, 1, oder 2 erwartet
    und diese als Integerwert zurückgibt.
    """
    
    # Der Wert der Eingabe wird als falscher Wert initialisiert
    eingabe = 'falsch'
    
    # Frage nach erneuter Eingabe, solange die Eingabe nicht 0, 1 oder 2 ist
    while eingabe not in ['0', '1', '2']:
        
        # Hier sollte nicht konvertiert werden, da wir sonst einen Fehler bei falschen Eingaben erzeugen
        eingabe = input("Wähle eine dieser Zahlen (0, 1, 2): ")
        
        if eingabe not in ['0', '1', '2']:
            # THIS ENTFERNT DIE DERZEITIGE AUSGABE DER ZELLE
            clear_output()
            
            print("Leider ist deine Eingabe nicht 0, 1 oder 2. Bitte versuche es erneut")
            
    
    # Zusätzlich könnten wir die Ausgabe nach erfolgreicher Nutzereingabe entfernen
    # clear_output()
    
    # Hier konvertieren wir, nachdem die while-Schleife erfolgreich beendet wurde
    return int(eingabe)
```

```python
def waehle_ersatz(spiel_liste, position):
    """ Aktualisiert spiel_liste mit einer Nutzereingabe am Index position. """
    
    nutzer_position = input("Gib den neuen Wert für die gewählte Position ein")
    
    spiel_liste[position] = nutzer_position
    
    return spiel_liste
```

```python
def spielstart_eingabe():
    """ Funktion, die vom Spieler erfragt, ob er ein Spiel beginnen möchte
    und einen dementsprechenden bool'schen Wert zurückgibt.
    """
    
    # Der Wert der Eingabe wird als falscher Wert initialisiert
    eingabe = 'falsch'
    
    # Erwarte J oder N als Eingabe
    while eingabe not in ['J','N']:
        
        eingabe = input("Möchtest du weiterspielen? J oder N")

        
        if eingabe not in ['Y','N']:
            # THIS ENTFERNT DIE DERZEITIGE AUSGABE DER ZELLE
            clear_output()
            
            print("Leider habe ich deine Eingabe nicht verstanden. Bitte antworte J oder N.")
            
    
    # Zusätzlich könnten wir die Ausgabe nach erfolgreicher Nutzereingabe entfernen
    # clear_output()
    
    
    if eingabe == "J":
        # Weiterspielen
        return True
    else:
        # Aufhören
        return False
```

**Die gesamte Spiellogik**

```python
# Variable zur Kontrolle des Spielstandes
spiel_laeuft = True

# Erste Spielliste
spiel_liste = [0, 1, 2]

while spiel_laeuft:
    
    # Entferne Ausgaben und zeige die aktuelle Spielliste
    clear_output()
    liste_ausgeben(spiel_liste)
    
    # Lass den Spieler eine Position wählen
    position = waehle_position()
    
    # Ersetze den Wert an der Position und aktualisiere spiel_liste
    spiel_liste = waehle_ersatz(spiel_liste, position)
    
    # Entferne Ausgaben und zeige die aktualisierte spiel_liste
    clear_output()
    liste_ausgeben(spiel_liste)
    
    # Frage, ob das Spiel weiter geht
    spiel_laeuft = spielstart_eingabe()
```
