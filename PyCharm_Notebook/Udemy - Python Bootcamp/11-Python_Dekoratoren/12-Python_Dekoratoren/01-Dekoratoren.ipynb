# Dekoratoren

Dekoratoren (engl. decorators) sind Python-Objekte, welche die Funktionalität anderer Funktionen oder Klassen verändern und ergänzen. Sie sind sogenannter *syntaktischer Zucker* und helfen dabei, Code kürzer und besser lesbar darzustellen.

Der erste Kontakt mit Dekoratoren ist häufig schwierig, weshalb wir mit einer kleine Wiederholung beginnen. Um Dekoratoren ausreichend zu erklären, werden wir langsam auf Funktionen aufbauen.

*Stelle sicher, vorher die Notebooks neu zu starten, damit die Ergebnisse identisch sind.*

## Funktionen Wiederholung

```python
def func():
    """ Einfache Demofunktion. Gibt 1 zurück. """

    return 1
```

```python
func()
```

## Geltungsbereich (Scope) Wiederholung

Erinnert euch an die Lektion zu verschachtelten Anweisungen und daran, dass Python den Geltungsbereich nutzt, um zu wissen, worauf sich ein Name bezieht. Zum Beispiel:

```python
mein_string = "Globale Variable"


def lokale_variablen():
    """ Funktion zur Ausgabe der lokalen Variablen. """

    print(locals())
```

Beachte, dass Funktionen einen neuen Geltungsbereich schaffen. D.h. die Funktionen haben ihren eigenen Namensbereich, den sie auf Namen überprüfen, wenn diese innerhalb der Funktion genannt werden. Wir können überprüfen, welche lokalen und globalen Variablen aktuell definiert sind, indem wir `locals()` und `globals()` verwenden. Zum Beispiel:

```python
print(globals())
```

Wir erhalten hier ein Dictionary zurück, dass alle globalen Variablen beinhaltet. Viele davon sind in Python vordefiniert. Lasst uns fortfahren und deren Keys überprüfen:

```python
print(globals().keys())
```

Ist euch aufgefallen, dass **mein_string** darunter ist, die Variable, die wir weiter oben definiert haben?

```python
globals()['mein_string']
```

Jetzt können wir unsere Funktion lokale_variablen() benutzen, um zu überprüfen, ob es lokale Variablen gibt. Es sollte keine geben...

```python
lokale_variablen()
```

Großartig! Jetzt können wir damit beginnen, die Logik eines Dekorators zu verstehen. Hier spielt es wieder eine Rolle, **dass in Python alles ein Objekt ist**. Das bedeutet, dass Funktionen Objekte sind, die benannt und an andere Funktionen übergeben werden können. Lasst uns mit einem einfachen Beispiel beginnen:

```python
def hallo(name='Rene'):
    """ Demofunktion. Gibt einen Gruss mit name zurück. """
    
    return 'Hallo ' + name
```

```python
hallo()
```

Jetzt können wir der Funktion einen anderen Namen geben. Beachte dass wir an dieser Stelle keine Klammern verwenden. Wir wollen die Funktion **hallo** nicht ausführen! Stattdessen platzieren wir sie nur in der Variablen `gruss`.

```python
gruss = hallo
```

```python
gruss
```

```python
gruss()
```

Was passiert also, wenn wir **hallo** löschen?

```python
del hallo
```

```python
hallo()
```

```python
gruss()
```

Obwohl wir **hallo** gelöscht haben, zeigt **gruss** *immer noch* auf das ursprüngliche Funktionsobjekt. So wurde gezeigt, dass Funktionsobjekte anderen Objekten zugewiesen werden können.

## Funktionen innerhalb von Funktionen

Toll! Wir haben gesehen, wie wir Funktionen als Objekte behandeln können. Jetzt können wir uns anschauen, wie man Funktionen innerhalb von anderen Funktionen definieren kann:

```python
def hallo(name="Rene"):
    """ Funktion zur Ausgabe eines Grusses mit zwei Unterfunktionen
    mit weiteren Ausgaben. 
    """
    
    print("Die hallo() Funktion wurde ausgeführt")
    
    def gruss():
        """ Funktion zur Demonstration von inneren Funktionen. """
        
        return '\t Das ist innerhalb der gruss() Funktion'
    
    def willkommen():
        """ Funktion zur Demonstration von inneren Funktionen. """
        
        return '\t Das ist innerhalb der willkommen() Funktion'
    
    print(gruss())
    print(willkommen())
    print("Jetzt sind wir wieder innerhalb der hallo() Funktion")
```

```python
hallo()
```

```python
willkommen()
```

Durch den Geltungsbereich der Funktionen ist `willkommen` außerhalb von `hallo` nicht erreichbar. Jetzt können wir lernen, wie wir auf Funktionen innerhalb von Funktionen zugreifen können.

## Funktionen zurückgeben

```python
def hallo(name="Rene"):
    """ Funktion zur Ausgabe eines Grusses mit zwei Unterfunktionen
    mit weiteren Ausgaben. 
    """
    
    def gruss():
        """ Funktion zur Demonstration von inneren Funktionen. """
        
        return '\t Das ist innerhalb der gruss() Funktion'
    
    def willkommen():
        """ Funktion zur Demonstration von inneren Funktionen. """
        
        return '\t Das ist innerhalb der willkommen() Funktion'
    
    if name == "Rene":
        return(gruss)
    else:
        return(wilkommen)
```

Lasst uns schauen, welche Funktion wir erhalten, wenn wir `neu = hallo()` setzen. Die geschlossenen Klammern bewirken, dass der Name als Rene definiert wurde.

```python
neu = hallo()
```

```python
neu
```

Gut! Jetzt können wir uns anschauen, wie `neu` zur Funktion `gruss` innerhalb der Funktion `hallo` weiterleitet.

```python
print(neu())
```

An dieser Stelle sollten wir uns noch einmal den Code anschauen.

In der `if/else`-Abfrage geben wir `gruss` und `willkommen` zurück, nicht `gruss()` und `willkommen()`. 

Das liegt daran, dass die Klammern hinter dem Namen dafür sorgen, dass die Funktion ausgeführt wird. Im Gegensatz dazu wird der Code nicht ausgeführt, wenn wir die Klammern weglassen und nur den Namen des Funktionsobjektes übergeben.

Wenn wir `neu = hallo()` schreiben, wird `hallo()` ausgeführt und weil der Name per Definition "Rene" ist, wird die Funktion `gruss` zurückgegeben. Wenn wir die Anweisung `neu = hallo(name="Sam")` eingeben, erhalten wir die Funktion `willkommen` zurück. Wir können ebenfalls `print(hallo()())` ausführen, was die Ausgabe *Das ist innerhalb der gruss() Funktion* erzeugt.

## Funktionen als Parameter

Lasst uns jetzt betrachten, wie wir Funktionen als Parameter an andere Funktionen übergeben:

```python
def hallo():
    """ Funktion zur Ausgabe eines einfachen Strings. """
    
    return "Hallo Rene!"

def andere(func):
    """ Funktion zur Demonstration der Annahme einer Funktion als Parameter."""
    
    print("Anderer Code würde hier stehen")
    print(func())
```

```python
andere(hallo)
```

Super! Achte darauf, wie wir die Funktion als Objekt übergeben und sie dann innerhalb einer anderen Funktion genutzt haben. Jetzt können wir unseren ersten Dekorator:

## Einen Dekorator erstellen

In dem vorherigen Beispiel haben wir tatsächlich schon manuell einen Dekorator erstellt. Hier werden wir ihn modifizieren, um die Anwendung klarer herauszustellen:

```python
def neu_dekorator(funk):
    """ Dekorator Demo äußere Funktion. """
    
    
    def huellen_funktion():
        """ Dekorator Demo innere (aufrufende) Funktion. """
        
        print("Code würde hier stehen, der vor der Funktion ausgeführt wird.")
        
        funk()
        
        print("Hier steht, was nach der Funktion ausgeführt wird.")
        
    return huellen_funktion

def func_braucht_dek():
    """ Dekorator Demo zu dekorierende Funktion. """
    
    print("Diese Funktion braucht einen Dekorator")
```

```python
func_braucht_dek()
```

```python
# Ersetze die Funktion func_braucht_dec mit der dekorierten Version
func_braucht_dek = neu_dekorator(func_braucht_dek)
```

```python
func_braucht_dek()
```

Was ist hier passiert? Der Decorator hat die Funktion umschlossen und ihr Verhalten geändert. Jetzt zeigen wir noch, wie wir diesen Code in Decorator-Schreibweise mit Hilfe des Symbols `@` schreiben können:

```python
@neu_dekorator
def func_braucht_dek():
    """ Dekorator Demo zu dekorierende Funktion. """
    
    print("Diese Funktion braucht einen Dekorator.")
```

```python
func_braucht_dek()
```

Jetzt haben wir manuell einen Dekorator erstellt und gesehen, wie wir das Symbol `@` in Python dafür verwenden können, um dies zu automatisieren und den Code aufzuräumen. Z.B. bei der Entwicklung von Webanwendungen, mit Flask oder Django wirst du sehr oft auf Dekoratoren treffen.
