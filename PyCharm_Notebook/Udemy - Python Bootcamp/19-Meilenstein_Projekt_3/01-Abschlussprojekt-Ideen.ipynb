# Liste von Projektideen

Hier ist eine Liste von etwa 100 Projektideen, die du in Python implementieren kannst. Manche davon sind sehr einfach, manche wurden bereits in früheren Notebooks vorgestellt (z.B. Fibonacci und FizzBuzz) und manche sind durchaus anspruchsvoll.

Nimm zuerst ein einfaches Projekt, dass du an einem Tag fertig bekommst, und nimm dir danach ein schwierigeres Projekt vor, für das du gegebenenfalls eine Woche mit Recherche im Internet brauchen könntest.

Verlinkt bei den Projekten sind auch Beispiellösungen, falls du nicht weiter kommst.

Sicherheit
-------------

**Cäsarverschlüsselung** - Implementiere die Cäsar-Verschlüsselung sowohl zum Ver- als auch Entschlüsseln. Bei dieser Verschlüsselung wird das Alphabet rotiert um **n** Stellen (Nutzereingabe). Dabei wird von Z zu A umgebrochen. So kodiert z.B. **n=2** "HI" als "JK" und **n=20** "HI" als "BC". Diese einfache "monoalphabetische Substitionsverschlüsselung" bietet so gut wie keine Sicherheit, da der Angreifer entweder Frequenzanalyse zum entschlüsseln einsetzen kann oder einfach alle 25 möglichen Verschiebungen ausprobiert. **Lösungen:** [[shyamsalimkumar (Python)]](https://github.com/shyamsalimkumar/Projects/blob/master/Security/caesar_cipher.py) [[sriniavireddy (python)]](https://github.com/sriniavireddy/scripts-and-solutions/blob/master/CaesarCipher.py)
