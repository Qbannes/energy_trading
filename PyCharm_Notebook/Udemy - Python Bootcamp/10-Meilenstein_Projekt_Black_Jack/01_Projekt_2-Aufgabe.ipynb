# Meilenstein-Projekt 2 - Blackjack
In diesem Meilenstein-Projekt wirst du ein vollständiges BlackJack-Spiel in Python implementieren.

Hier sind die Anforderungen:

* Du sollst ein vollständiges, aber einfaches [BlackJack](https://de.wikipedia.org/wiki/Black_Jack)-Spiel implementieren
* Das Spiel soll einen menschlichen Spieler gegen einen Computergegner beinhalten
* Der Spieler kann halten oder nehmen
* Der Spieler soll den Wettbetrag eingeben können
* Der Betrag des Spielergeldes soll über den Spielverlauf aktualisiert werden
* Der Spieler soll über das Spielgeschehen informiert werden (Gewinn, Verlust, Bust, ...)

Und besonders wichtig:

* **Du musst OOP und Klassen in einigen Teilen deines Spieles verwenden. Funktionen alleine sind nicht ausreichend. Verwende Klassen für den Kartenstapel und die Spielerhand. Es gibt viele Wege, dies korrekt zu implementieren - finde deinen eigenen Weg!**
