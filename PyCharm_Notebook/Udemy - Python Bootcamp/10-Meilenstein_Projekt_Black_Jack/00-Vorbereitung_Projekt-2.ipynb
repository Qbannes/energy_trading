## Kartenspiel War

Lasst uns gemeinsam zuerst ein einfaches, in OOP implementiertes Kartenspiel namens War entwickeln, bevor wir uns dem komplexeren Meilenstein Projekt 2 stellen. In diesem Spiel starten zwei Spieler mit der Hälfte des Kartenstapels. Dann entfernt jeder Spieler eine Karte aus seinem Stapel, die Karten werden verglichen und der Spieler mit der höherwertigen Karte gewinnt beide Karten und fügt sie seinem Stapel hinzu. Falls die Karten den gleichen Wert haben, startet der Gefechtsmodus. Im Gefechtsmodus verwirft jeder Spieler die nächsten drei Karten und zieht die vierte. Die stärkere Karte gewinnt und alle Karten werden dem Gewinner hinzugefügt. Falls die Karten gleich sind, bleibt das Spiel im Gefechtsmodus. Hat ein Spieler zu wenige Karten, so verliert dieser das Spiel.

Das Spiel ist nur eine Demonstration ohne Nutzerinteraktion!

## Klasse Karte

### Erzeugen einer Klasse Karte mit globalen Konstanten

Hier definieren wir ein paar global verfügbare Konstanten, die sich im Spiel nicht ändern werden. Beispielsweise die Karte-Wert-Zuordnung ändert sich im Verlauf des Spieles nicht.

```python
# Das brauchen wir später
import random 
```

```python
FARBEN = ('Herz', 'Karo', 'Pik', 'Kreuz')
RAENGE = ('Zwei', 'Drei', 'Vier', 'Fuenf', 'Sechs', 'Sieben',
          'Acht', 'Neun', 'Zehn', 'Bube', 'Dame', 'Koenig', 'Ass')
WERTE = {'Zwei': 2, 'Drei': 3, 'Vier': 4, 'Fuenf': 5, 'Sechs': 6, 'Sieben': 7, 'Acht': 8,
         'Neun': 9, 'Zehn': 10, 'Bube': 11, 'Dame': 12, 'Koenig': 13, 'Ass': 14}
```

```python
class Karte:
    """ Klasse für eine einzelne Spielkarte mit farbe, rang und wert sowie
    einer Stringrepräsentation. 
    """

    def __init__(self, farbe, rang):
        self.farbe = farbe
        self.rang = rang
        self.wert = WERTE[rang]

    def __str__(self):
        return self.farbe + '-' + self.rang
```

## Klasse Stapel

### Eine Klasse in einer Klasse verwenden

Wir haben gerade eine Einzelkarte erzeugt, aber wie können wir einen kompletten Kartenstapel erstellen? Lasst uns das mithilfe der Klasse Karte untersuchen.

Ein Kartenstapel besteht aus vielen Karten, was bedeutet, dass wir die Klasse Karte in der `__init__`-Methode von Kartenstapel verwenden werden.

```python
class Stapel:
    """ Klasse Stapel erzeugt bei der Initialisierung einen Kartenstapel für
    alle Farben und Ränge und stellt Methoden zum mischen und austeilen von 
    Karten bereit.
    """

    def __init__(self):
        """ Initialisierung der Klasse Stapel. Karten aller Farben und Ränge
        werden erzeugt. Die Funktion wird nur einmal, und zwar bei der 
        Instanziierung von Stapel, ausgeführt. 
        """

        self.stapel = []  # Beginne mit einer leeren Liste

        for farbe in FARBEN:
            for rang in RAENGE:
                # Erzeuge Karte-Objekte und füge sie der Liste hinzu
                self.stapel.append(Karte(farbe, rang))

    def mischen(self):
        """ Methode zum mischen des gesamten Kartenstapels"""

        random.shuffle(self.stapel)

    def austeilen(self):
        """Diese Methode entfernt die letzte Karte im Stapel und gibt sie zurück. """

        einzelkarte = self.stapel.pop()
        return einzelkarte
```

# Klasse Spieler

Als nächstes implementieren wir die Klasse Spieler. Ein Spieler sollte dazu in der Lage sein, Karteninstanzen halten sowie anzunehmen und zu entfernen. Die Klasse Spieler sollte flexibel genug sein, um eine einzige oder mehrere Karten gleichzeitig anzunehmen, damit wir nicht die gleiche Methode wiederholt aufrufen müssen.

Wir behalten das alles im Hinterkopf beim entwickeln der Klasse Spieler.

```python
class Spieler:

    def __init__(self, name):
        """ Initialisierung mit Spielernamen und leerem Kartenstapel. """

        self.name = name
        self.stapel = []

    def entferne_eine(self):
        """ Entferne die oberste Karte aus dem Spielerstapel und gib sie zurück. """

        # Wir entfernen eine Karte des Spielerstapels
        # Mit Index 0 bei pop entfernen wir von 'oben' aus dem Stapel
        # Wir nehmen an, dass -1 die unterste Karte im Stapel ist
        return self.stapel.pop(0)

    def karten_hinzufuegen(self, neue_karten):
        """ Nimm entweder eine einzelne Karte oder eine Liste von Karten an und 
        füge sie dem Spielerstapel hinzu.
        """

        if type(neue_karten) == type([]):
            self.stapel.extend(neue_karten)
        else:
            self.stapel.append(neue_karten)

    def __str__(self):
        """ Stringrepräsentation des Spielers. """
        
        return f'Spieler {self.name} hat {len(self.stapel)} Karten.'
```

## Spiele das Spiel

```python
import pdb
```

```python
spiel_starten = True
```

```python
anzahl_runden = 0

while spiel_starten:

    anzahl_runden += 1
    print(f"Runde {anzahl_runden}")

    # Überprüfe, ob ein Spieler keine Karten mehr hat
    if len(spieler_eins.stapel) == 0:
        print("Spieler Eins hat keine Karten mehr! Game Over")
        print("Spieler Zwei gewinnt!")
        spiel_starten = False
        break

    if len(spieler_zwei.stapel) == 0:
        print("Spieler Zwei hat keine Karten mehr! Game Over")
        print("Spieler Eins gewinnt!")
        spiel_starten = False
        break

    # Andernfalls geht das Spiel weiter!

    # Starte eine neue Runde und setze die Karten "auf dem Tisch" zurück
    spieler_eins_karten = []
    spieler_eins_karten.append(spieler_eins.entferne_eine())

    spieler_zwei_karten = []
    spieler_zwei_karten.append(spieler_zwei.entferne_eine())

    im_gefecht = True

    while im_gefecht:

        # print('Runnning')

        if spieler_eins_karten[-1].wert > spieler_zwei_karten[-1].wert:

            # Spieler Eins bekommt die Karten
            spieler_eins.karten_hinzufuegen(spieler_eins_karten)
            spieler_eins.karten_hinzufuegen(spieler_zwei_karten)

            # Nicht mehr "im Gefecht", Zeit für die nächste Runde
            im_gefecht = False

        # Spieler Zwei hat die höhere Karte
        elif spieler_eins_karten[-1].wert < spieler_zwei_karten[-1].wert:

            # Spieler Zwei bekommt die Karten
            spieler_zwei.karten_hinzufuegen(spieler_eins_karten)
            spieler_zwei.karten_hinzufuegen(spieler_zwei_karten)

            # Nicht mehr "im Gefecht", Zeit für die nächste Runde
            im_gefecht = False

        else:
            # Entferne solange drei Karten und spiele die vierte, bis eine
            # hoehere Karte gespielt wurde oder ein Spieler zuwenig Karten hat
            while spieler_eins_karten[-1].wert == spieler_zwei_karten[-1].wert:
                print('Gefechtsmodus !')
                if len(spieler_eins.stapel) < 4:
                    print(
                        "Spieler Eins kann das Gefecht nicht weiter führen! Game Over!")
                    print("Spieler Zwei gewinnt!")
                    spiel_starten, im_gefecht = False, False
                    break

                elif len(spieler_zwei.stapel) < 4:
                    print(
                        "Spieler Zwei kann das Gefecht nicht weiter führen! Game Over!")
                    print("Spieler Eins gewinnt!")
                    spiel_starten, im_gefecht = False, False
                    break

                else:
                    for num in range(4):
                        spieler_eins_karten.append(
                            spieler_eins.entferne_eine())
                        spieler_zwei_karten.append(
                            spieler_zwei.entferne_eine())

            # Die zuletzt gespielten Karten sind verschieden,
            # beide Spieler haben noch Karten
            if im_gefecht == True:
                im_gefecht = False
                if spieler_eins_karten[-1].wert > spieler_zwei_karten[-1].wert:
                    spieler_eins.stapel += spieler_eins_karten
                    spieler_eins.stapel += spieler_zwei_karten
                else:
                    spieler_zwei.stapel += spieler_eins_karten
                    spieler_zwei.stapel += spieler_zwei_karten
```
