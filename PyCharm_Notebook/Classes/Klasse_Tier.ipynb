```python
# Vererbung in Python

class Tier:   # Klasse Tier
    rasse=''  # Atribute Rasse Geschlecht und Alter
    geschlecht=''
    alter=0

    def __init__(self, rasse, geschlecht, alter): # Konstruktor
        self.rasse=rasse
        self.geschlecht=geschlecht
        self.alter=alter

    def printRasse(self):
        print('Rasse: ',self.rasse)


class Hund(Tier): # Klasse Hunde erbt von Klasse Tier die Atribute
    geruchsinn='sehr gut'

    def __init__(self, rasse, geschlecht, alter):
        super(Hund,self).__init__(rasse, geschlecht, alter) # Aufrufen
        self.geruchsinn=geruchsinn

h1=Hund('sehr gut','bluthund','männlich')# einen Hund erstellen

h1.printRasse()
```

```python
# Vererbung in Python

class Tier:   # Klasse Tier
    rasse=''  # Atribute Rasse Geschlecht und Alter
    geschlecht=''
    alter=0

    def __init__(self, rasse, geschlecht, alter): # Konstruktor
        self.rasse=rasse
        self.geschlecht=geschlecht
        self.alter=alter

    def printRasse(self):
        print('Rasse: ',self.rasse)


class Hund(Tier): # Klasse Hunde erbt von Klasse Tier die Attribute
    geruchsinn='sehr gut'

    def __init__(self, rasse, geschlecht, alter, geruchssinn):  # ← GEÄNDERT: Parameter für geruchssinn hinzugefügt
        super(Hund,self).__init__(rasse, geschlecht, alter) # Aufrufen
        self.geruchsinn=geruchssinn  # ← GEÄNDERT: Parameter 'geruchssinn' statt undefiniertem 'geruchsinn'

# GEÄNDERT: Korrekte Reihenfolge + geruchssinn-Argument (Standardwert 'sehr gut' möglich)
h1=Hund('bluthund','männlich',5,'sehr gut')# einen Hund erstellen

h1.printRasse()
```
