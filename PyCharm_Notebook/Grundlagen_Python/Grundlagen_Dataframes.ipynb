# Teil 2

```python
# Dataframe anlegen
# Beim Index müssen die Werte nicht separat übergeben werden ('a','b','c','d'). 
# Es genügt einfach 'a b c d' zu schreiben, muss aber mit split() wieder unterteilt werden.

df = pd.DataFrame(randn(5,4), index='A B C D E'.split(),columns='W X Y Z'.split())
```

```python
# bedingte Auswahl:
# Boolianwerte ausgeben

df>0
```

```python
# bedingte Auswahl

df[df>0]
```

```python
# Verschärfte bedingte Auswahl

df['W']>0
```

```python
df[df['W']>0]
```

```python
df[df['W']>0]['Y']
```

```python
df[df['W']>0][['Y','X']]
```

```python
df[df['W']>0][['Y','X','Z']]
```

```python
# Neues Dataframe anlegen: Bedingungen müssen mit runder Klammer übergeben werden!
# Das 'and' führt hier zum Fehler ambiguous=mehrdeutig

df[(df['W']>0)and(df['Y']>1)]
```

```python
# anstatt das and muss & verwendet werden zum erstellen des neuen Dataframes

df[(df['W']>0)&(df['Y']>1)]
```

```python
# anstatt ein or hier ein | verwenden zum erstellen des neuen Dataframes

df[(df['W']>0)|(df['Y']>1)]
```

```python
df
```

```python
# Index auf Standarindex zurücksetzen

df.reset_index()
```

```python
# Index wieder neue Werte zuweisen

neuind = 'BY BW HH BB NW'.split()
```

```python
df['Staaten']=neuind
```

```python
df
```

```python
# Index ändern: Aber wird nicht gespeichert! Probiere separate Ausgabe

df.set_index('Staaten')
```

```python
df
```

```python
# Index dauerhaft ändern mit inplace=True

df.set_index('Staaten', inplace=True)
```

```python
df
```

# Teil 3
- Multiindexierung

```python
aussen = ['G1','G1','G1','G2','G2','G2']
innen = [1,2,3,1,2,3]
hier_index = list(zip(aussen, innen))
hier_index = pd.MultiIndex.from_tuples(hier_index)
```

```python
hier_index
```

```python
df = pd.DataFrame(np.random.randn(6,2), index=hier_index,columns=['A','B'])
```

```python
df
```

```python
df.loc['G1'].loc[1]
```

```python
df.index.names
```

```python
df.index.names = ['Gruppe', 'Num']
```

```python
df
```

```python
df.xs('G1')
```

```python
df.xs(('G1',1))
```

```python
df.xs(1,level='Num')
```
