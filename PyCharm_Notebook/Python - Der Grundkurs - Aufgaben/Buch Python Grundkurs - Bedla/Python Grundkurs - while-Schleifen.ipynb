```python
# Als Alternative zu for können Sie Schleifen mit while formulieren. Die
# eingerückten Anweisungen werden dann so lange ausgeführt, wie die
# Bedingung erfüllt ist. Die Schlüsselwörter break, continue und else funktionieren wie bei der for-Schleife.

while bedingung:
    anweisungen
```

```python
# Im folgenden Beispiel durchläuft die Schleifenvariable die Werte von 1 bis
# 4. Am Ende hat i den Wert 5. Dieser Wert wird aber nicht mehr ausgegeben,
# weil die Schleife dann beendet wird.
i=1
while i < 5:
    print(i)
    i += 1
# Ausgabe 1 2 3 4
```

```python
# Vorsicht, Endlosschleife!
# Bei der Programmierung von Schleifen passiert es leicht, dass diese
# endlos laufen. Das wäre z. B. der Fall, wenn Sie im obigen Code die Anweisung i+=1 vergessen oder die Schleifenbedingung falsch formulieren.

i=1
while i:
    print(i)
    i += 1
# Ausgabe 1 bis unendlich
```

```python
# while-Schleifen sind insbesondere dann vorteilhaft, wenn die Zahlen unregelmäßige Abstände aufweisen sollen und eine Konstruktion mit for
# und range daher nicht zielführend ist:
i=1
while i<10000:
    print(i, end= ' ')
    i += i * i
# Ausgabe: 1 2 6 42 1806
# Hinweis: Der Parameter end=' ' in der print-Funktion legt fest, dass nach der Ausgabe ein Leerzeichen und keine neue Zeile angehängt wird.
```

```python
# Seite 145:Beispiele für Schleifen

# Beispiele für Schleifen, sowohl in der »klassischen« Syntax als auch in
# der Comprehension-Form, werden Sie in nahezu allen weiteren Kapiteln
# finden. Dennoch möchte ich Ihnen schon an dieser Stelle einige Anwendungsmöglichkeiten zeigen, bevor Sie Ihr Wissen im nächsten Abschnitt
# anhand einiger Übungen überprüfen können.

# 9 Verzweigungen und Schleifen

# Summe der Zahlen von 1 bis 1000
# Der Mathematiker Carl Friedrich Gauß entdeckte schon im Alter von neun
# Jahren, dass die Summe der Zahlen von 1 bis n mit dem folgenden Ausdruck berechnet werden kann:
# summe = n * (n + 1) / 2

# Aber wenn Sie diese Formel nicht kennen, können Sie natürlich auch den
# Computer rechnen lassen. In Python gibt es dafür viele Möglichkeiten. Am
# naheliegendsten ist vermutlich eine »gewöhnliche« for-Schleife, wobei
# der zu durchlaufende Zahlenraum mit range formuliert wird.
# Die Summe wird schrittweise in der Variablen sum errechnet. Dabei ist sum
# += i die Kurzschreibweise für den Ausdruck sum = sum + i (siehe Kapitel 4,
# »Operatoren«).

# Beispieldatei summe.py
sum = 0
for i in range(1, 1001):
    sum += i
print( 'Summe der Zahlen von 1 bis 1000: ', sum)
# Ergebnis: 500500
```

```python
# Längere Schreibweise

sum = 0
for i in range(1, 1001):
    sum = sum + i
print( 'Summe der Zahlen von 1 bis 1000: ', sum)
# Ergebnis: 500500
```

```python
# Wenn Sie mit while arbeiten, benötigen Sie zwei Variablen: Eine Schleifenvariable (hier i) und eine zur Berechnung der Summe:
sum = 0
i=1
while i <= 1000:
    sum += i
    i += 1

print( 'Summe der Zahlen von 1 bis 1000: ', sum)
# Ergebnis: 500500
```

```python
# Es geht aber auch ganz ohne (sichtbare) Schleife: Mit list(range(...)) bilden Sie eine Liste mit den Zahlen von 1 bis 1000. reduce wendet nun die
# Formel x+y auf die ersten beiden Listenelemente an, dann auf das Zwischenergebnis und das dritte Listenelement usw. (Intern arbeitet reduce
# also ebenfalls wie eine Schleife.)

from functools import reduce
sum = reduce(lambda x, y: x + y, list(range(1, 1001)))
print(sum)
# Ergebnis: 500500
```

```python
# Zurück zur Schleife: Wenn Sie wissen, dass Python die Funktion sum zur
# Verfügung stellt, um die Elemente einer Liste oder eines Generators zu
# addieren, ergibt sich eine weitere Variante:

sum = sum( i for i in range(1, 1001))
```

```python
# Einmaleins-Tabelle
# Stellen Sie sich vor, Sie wollten mit einem Python-Script eine Tabelle mit
# allen Multiplikationen der Zahlen zwischen 1 und 10 ausgeben, die wie
# folgt aussieht:
# 1x1=1 2x1=2 3x1=3 4x1=4 ... 10x1=10
# 1x2=2 2x2=4 3x2=6 4x2=8 ... 10x2=20
# 1x3=3 2x3=6 3x3=9 4x3=12 ... 10x3=30
# ...
# 1x10=10 2x10=20 3x10=30 4x10=40 ... 10x10=100
# Für derartige Aufgabenstellungen benötigen Sie zwei Schleifen, die ineinander angeordnet sind:
# # Beispieldatei einmaleins.py
for i in range(1, 11):
    for j in range(1, 11):
        print( '%dx%d=%d\t ' % (j, i, i*j), end= ' ')
    print()

# Dabei beginnt die äußere Schleife mit i=1. In der inneren Schleife durchläuft jetzt j alle Werte von 1 bis 10. Dabei wird die erste Zeile der Multiplikationstabelle ausgegeben. Wenn die j-Schleife fertig ist, geht es in der
# i-Schleife weiter, und es gilt i=2. Die j-Schleife wird nun erneut ausgeführt,
# wieder von 1 bis 10. So entsteht die zweite Zeile usw.
# %d in der Formatierungszeichenkette von print wird durch die drei Ausdrücke j, i und i*j ersetzt. \t fügt der Ausgabe ein Tabulatorzeichen hinzu.
# end='' verhindert, dass nach jeder Ausgabe eine neue Zeile beginnt. Eine
# neue Zeile soll erst beginnen, wenn die j-Schleife fertig ist. Die Anweisung
# print() ist deswegen so eingerückt, damit sie Teil der i-Schleife ist, also mit
# jedem Durchlauf der i-Schleife ausgeführt wird.
```

```python
# Parameter verarbeiten
# Beim Aufruf eines Python-Scripts können Sie Parameter übergeben. Im
# Script enthält sys.argv aus dem sys-Modul zuerst den Programmnamen
# und dann die Parameter in Form einer Liste.
# argv ist eine Liste, deren erstes Element der Dateiname des Scripts ist. Die
# weiteren Listenelemente enthalten die übergebenen Parameter. Da der
# Script-Name selten benötigt wird, wird er mit dem Ausdruck [1:] eliminiert.

import sys
if(len(sys.argv) <= 1):
    print( 'Es wurden keine Parameter übergeben. ')
else:
    for x in sys.argv[1:]:
        print( 'Parameter:', x)
```

```python
# Unter Windows funktioniert das nicht. Dort müssen Sie die glob-Funktion
# aus dem gleichnamigen Modul einsetzen: glob('*.txt') liefert eine Liste
# mit allen Textdateien im gerade aktuellen Verzeichnis. (Beachten Sie, dass
# diese Liste leer ist, wenn es keine passenden Dateien gibt! Unter macOS
# oder Linux bleibt in solchen Fällen *.txt als Parameter erhalten.)
# Mit der Berücksichtigung dieses Sonderfalls bekommen wir plötzlich ein
# schönes Beispiel für die Verschachtelung von Bedingungen und Schleifen
# und für die Konstruktion komplexer Bedingungen: Die erste if-Abfrage
# testet, ob überhaupt Parameter übergeben wurden. Ist das der Fall, verarbeitet die for-Schleife alle Parameter. Wenn in einem Parameter die
# Zeichen ? oder * auftauchen und das Script unter Windows läuft, wird die
# glob-Funktion ausgeführt, andernfalls einfach der Parameter angezeigt.

# Beispieldatei parameter.py
import platform , sys
from glob import glob
iswin = platform.system() == 'Windows '
if len(sys.argv) <= 1:
    print( 'Es wurden keine Parameter übergeben. ')
else:
    for para in sys.argv[1:]:
        if iswin and ( '?' in para or '* ' in para):
            print( 'Parameter:', glob(para))
    else:
        print( 'Parameter:', para)
```

```python
# 9.6 Pattern Matching
# In Python 3.10 wurde der Mechanismus des Pattern Matchings eingeführt.
# Damit können Sie die Elemente eines Tupels oder einer Liste verarbeiten.
# Nehmen Sie an, eine Liste enthält in ein, zwei oder drei Elemente die Kommandos quit, do <cmd> oder join <part1> <part2>. Anstatt zuerst zu testen,
# wie viele Listenelemente vorliegen, und diese dann umständlich auszuwerten, können Sie in Python 3.10 den folgenden Code verwenden:

#!/usr/bin/env python3
lst = ['join', 1, 2]
match lst:
    case ['quit']:
        print('Programmende')     # Nur bei exakt ['quit']
    case ['do', cmd]:
        print('Führe', cmd, 'aus.')   # Bei ['do', irgendwas]
    case ['join', t5, ppart2]:
        print('Füge', t5, 'und', ppart2, 'aneinander.') # Bei ['join', x, y]
    case _:
        print('Unbekanntes Kommando') # Wenn nichts passt
# Ausgabe: Füge 1 und 2 aneinander.

# Die prinzipielle Syntax geht aus dem Beispiel hervor: An match übergeben
# Sie eine Liste, ein Tupel oder eine andere Aufzählung von Daten. In der
# Folge geben die case-Ausdrücke vor, welchem Muster die Daten entsprechen müssen. case ['join', part1, part2] bedeutet, dass es sich um eine
# Liste mit drei Elementen handeln muss. Das erste Element der Liste muss
# die Zeichenkette 'join' sein. Für die restlichen beiden Elemente gibt es
# keine Regeln. Sie können im nachfolgenden Codeblock über die Variablen
# part1 und part2 verarbeitet werden.

# Bei der Verarbeitung von match werden der Reihe nach alle case-Muster
# ausgewertet, bis das erste zutrifft. Mit der Verarbeitung des nachfolgenden Codeblocks endet der gesamte match-Block. case _ eignet sich als
# abschließendes, immer zutreffendes case-Muster, um einen Defaultcodeblock einzuleiten.
```

```python
# case-Muster können mit zusätzlichen Bedingungen versehen werden, die
# in der Dokumentation als Guards bezeichnet werden:

lst = ['join', 1, 2]

match lst:
    case ['join', part1, part2] if part1 < part2:
        print(part1, 'ist kleiner als', part2)
    case _:
        print('Kein Treffer')

```

```python
# Beispiel 1: Protokollierte Befehle mit variabler Anzahl Argumente

command = ['join', 'User1', 'User2', 'User3']

match command:
    case ['join', *users]:
        print('Füge folgende Nutzer zusammen:', users)
    case _:
        print('Unbekanntes Kommando')
# Ausgabe: Füge folgende Nutzer zusammen: ['User1', 'User2', 'User3']
# Hier kannst du beliebig viele Nutzer übergeben, und das Pattern verarbeitet sie alle.
```

```python
# Beispiel 2: Verarbeitung von variablen Parameterlisten

def handle_args(args):
    match args:
        case ['print', *texts]:
            for text in texts:
                print(text)
        case _:
            print("Ungültige Eingabe")

handle_args(['print', 'Hallo', 'Welt', '!'])

# In der Praxis eignet sich *name hervorragend, wenn Funktionen oder Befehle mit einer variablen Anzahl
# von Argumenten oder Listenelementen rechnen müssen, ohne die Anzahl der Elemente vorher zu kennen.


# Fazit
# *name ermöglicht flexibles Pattern Matching mit beliebig vielen weiteren Elementen.

# Es entkoppelt die Struktur vom festen Umfang und macht Programme anpassungsfähiger und lesbarer.

# Besonders praktisch bei Eingabelisten oder sequenziellen Daten mit dynamischer Länge.
```

```python
# Lösung W7: Formulieren Sie eine Schleife, um den Wertebereich zwischen 125
# und 160 in elf Schritten zu durchlaufen. Das Programm soll alle elf Zahlen ausgeben, beginnend mit 125,0 und endend mit 160,0.



# 1. Wertebereich und Schrittweite festlegen

# Startwert: 125.0
# Endwert: 160.0
# Anzahl gewünschter Werte: 11

first_value = 125
last_value = 160
steps = 11

step_size = (last_value - first_value) / (steps - 1)

# 2. Schleife schreiben, die Index von 0 bis 10 (für 11 Werte) durchläuft
# Für jeden Index i berechner den aktuellen Wert als: value i = first_value + i*step_size

for i in range(steps):
    current_value = first_value + i*step_size
    print(current_value)
```
