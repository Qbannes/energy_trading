```python
transa_df = pd.DataFrame(api.get_transactions())
# Datumsformatierung der Spalte time und Zuweisung von Datum und Uhrzeit aus der Spalte time den neuen Spalten datum und uhrzeit
transa_df['time'] = pd.to_datetime(transa_df['time'], utc=True)
transa_df['datum'] = transa_df['time'].dt.date
transa_df['uhrzeit'] = transa_df['time'].dt.time

# Spalte 'time' entfernen
transa_df = transa_df.drop(columns=['time'])
# Spalten neu anordnen: datum an Index 2, uhrzeit an Index 3
cols = list(transa_df.columns)
# Entfernen der neuen Spalten aus der Liste, falls sie nicht am gewünschten Platz sind

cols.remove('datum')
cols.remove('uhrzeit')
cols.remove('alias')
cols.remove('homeCurrency')
cols.remove('accountUserID')
cols.remove('accountNumber')
cols.remove('siteID')
cols.remove('divisionID')
cols.remove('fundingReason')
cols.remove('financing')
cols.remove('commission')
cols.remove('guaranteedExecutionFee')
cols.remove('distance')
cols.remove('triggerCondition')
cols.remove('tradeID')
cols.remove('positionFinancings')
cols.remove('shortPositionCloseout')
cols.remove('longPositionCloseout')
cols.remove('stopLossOnFill')
cols.remove('tradeClose')
cols.remove('tradesClosed')
cols.remove('tradeOpened')
cols.remove('requestID')
cols.remove('timeInForce')
cols.remove('fullPrice')
# Einfügen an Indexposition 2 und 3
cols.insert(2, 'datum')
cols.insert(3, 'uhrzeit')
# Geänderte Spalten dem DataFrame zuweisen
transa_df=transa_df[cols]
# DataFrame nach Spalte 'datum' absteigend sortieren
transa_df = transa_df.sort_values(by='datum', ascending=False)
# Optional: Index zurücksetzen, damit die Zeilennummern durchgehend sind
# transa_df = transa_df.reset_index(drop=True)

transa_df.to_csv('Transaktionsdaten.csv', sep=';', index=False)
pd.read_csv('Transaktionsdaten.csv', sep=';')
```
