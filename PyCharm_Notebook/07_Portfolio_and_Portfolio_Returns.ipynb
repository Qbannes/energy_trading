# Portfolio of Assets and Portfolio Returns

```python
import pandas as pd
import numpy  as np
```

```python
prices=pd.DataFrame(data={'Asset_A':[100,112],'Asset_B':[100,104]})
prices
```

```python
prices['Total']=prices.Asset_A+prices.Asset_B
prices
```

```python
returns=prices.pct_change() # simple returns
returns
```

**Portfoliorendite (portfolio return)==Gewichteter Durchschnitt (50%=0.5) einfacher Rendite (weighted average of simple returns)**


```python
0.5*0.12+0.5*0.04
```

```python
log_returns=np.log(prices/prices.shift()) # Log returns
log_returns
```

```python
0.5*log_returns.iloc[1,0]+0.5*log_returns.iloc[1,1]
```

# **Take Home: While Log Return are time-additive, they are not asset-additive.**
(While simple returns are asset-additive, they are not time-additive)
