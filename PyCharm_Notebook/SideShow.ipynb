```python
import numpy as np
import time


max_attempts=8
success=False
wait=3 # initial waiting period: 3 second
wait_increase=5 # waiting period increase in seconds
attempt=0
while True:
    try:
        result=1/np.random.randint(0,1)
    except Exception as e:
        print(e,end='  /  ')
    else:
        print(result,end='  |  ')
        success=True
        break # stop when it worked
    finally:
        attempt+=1
        print(f'Attempt: {attempt}')
        if success==False:
            if attempt>=max_attempts: # if max_attempts is reached
                print('Programm stopped: max_attempts reached!')
                break
            else: # if we haven´t reached max_attempts, waitt before we start the next try
                time.sleep(wait)
                wait+=wait_increase # increase waiting time
```

```python
# PV with vectorized Numpy Code
#
# Calculate the PV of the following cashflows with vectorized Numpy Code. Print the final PV!
#
# - 300 USD in t1,
#
# - 500 USD in t2,
#
# - 1000 USD in t3,
#
# - 2000 USD in t4
#
# The required rate of return (Cost of Capital) is 6% p.a.

import numpy as np

cf = np.array([300, 500, 1000, 2000])

```

```python
import numpy as np
cf = np.array([300, 500, 1000, 2000])  # Cashflows in t=1 bis t=4
r = 0.06  # Required rate of return
t = np.arange(1, 5)  # Zeiträume
pv_factors = (1 + r) ** -t  # Diskontierungsfaktoren
pv = np.sum(cf * pv_factors)  # Vektorisierte PV-Berechnung
print(pv)
```
