```python
my_list=[1,2,3,'four',5,6,7]
```

```python
for i in my_list:
    try:
        result=10+i
    except Exception as e:
        print(f'{i} is not a number', end='  |  ')
        print(e,end='  |  ')
        #break
    else: # only if no error occured
        print(result,end='|')
        print(f'{i} is not a valid input', end='  |  ')
    finally: # in any case
        print('Valid or not valid, it doesn´t matter!')
```

```python
import numpy as np
import time


max_attempts=8
success=False
wait=3 # initial waiting period: 3 second
wait_increase=5 # waiting period increase in seconds
attempt=0
while True:
    try:
        result=1/np.random.randint(0,1)
    except Exception as e:
        print(e,end='  /  ')
    else:
        print(result,end='  |  ')
        success=True
        break # stop when it worked
    finally:
        attempt+=1
        print(f'Attempt: {attempt}')
        if success==False:
            if attempt>=max_attempts: # if max_attempts is reached
                print('Programm stopped: max_attempts reached!')
                break
            else: # if we haven´t reached max_attempts, waitt before we start the next try
                time.sleep(wait)
                wait+=wait_increase # increase waiting time
```
