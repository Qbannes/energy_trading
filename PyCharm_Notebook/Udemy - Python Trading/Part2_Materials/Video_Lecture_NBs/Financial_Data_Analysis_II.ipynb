```python
import pandas as pd
import matplotlib.pyplot as plt
pd.options.display.float_format='{:.4f}'.format
```

```python
close = pd.read_csv("close.csv", index_col = "Date", parse_dates = ["Date"])
close
```

```python
# In der Datei close.csv, in der Spalte BA die NaN-Werte eliminieren (dropna()) 
# und anschließend in ein DataFrame umwandeln (to_frame()) und zuletzt eine Kopie erstellen
# diese Kopie noch eienr Variable zuweisen

msft = close.MSFT.dropna().to_frame().copy() 
msft
```

```python
msft.dtypes
```

```python
# Spalte dauerhaft neu benennen, rename colum permanently in price

msft.rename(columns={'MSFT':'Price'},inplace=True) 
msft
```

```python
# verschiebt alle Werte um 1 nach unten und hinterlässt in der ersten Zeile
# ein NaN aber speichert die Änderung nicht inplace. Dazu müsste die Änderung mit = einer
# Variablen zugewiesen werden.
msft.shift(periods=1)
```

```python
# Mit msft['P-lag1'] wird eine eingefügt, der gleichzeitig die Kopie Spalte Price,
# in der um eins verschobenen Werte nach unten zugewiesen wird. Das Ergebnis ist dauerhaft gespeichert als msft

msft["P_lag1"] = msft.shift(periods = 1)
# msft["P_lag1"] = msft["Price"].shift(periods=1) # Gleicehr Effekt

msft
```

```python
# ALternative 1: flexible Subtraktion

msft['P-Differenz']=msft.Price.sub(msft.P_lag1)
```

```python
msft
```

```python
msft['P-Differenz'].head()
```

```python
# ALternative 2: Reihensubtraktion: Subtrahiert vorherigen Wert

msft["P_diff2"] = msft.Price.diff(periods = 1)
msft
```

```python
# Prüfung, ob die beiden Spalten P-Differenz und P-diff2 genau gleich sind, da sie mit unterschiedlichen Funktionen berechnet wurden
# Info: Aufgrund der Schreibweise von P-Differenz werden hier die Klammern benutzt, um die Spalte auszuwählen
msft['P-Differenz'].equals(msft.P_diff2)
```

```python
# Devision zweier Spalten minus 1 = relative Preisänderung

msft.Price.div(msft.P_lag1)-1
```
