# Investment Multiple and CAGR (Compund Annual Growth Rate)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
pd.options.display.float_format='{:.4f}'.format

plt.style.use('seaborn-v0_8')
#print(plt.style.available)
```

```python
msft=pd.read_csv('msft.csv')
msft
```

```python
msft=pd.read_csv('msft.csv', index_col='Date', parse_dates=True)
msft
```

# Investment Multiple: Ending Value of 1 Dollar invested.
Multiple = Ending Value / Initial Investment

```python
multiple=(msft.Price.iloc[-1]/msft.Price.iloc[0])
multiple
```

```python
(multiple-1)*100
# An increase of 519.0410313769353 %
```

```python
msft.Price/msft.Price.iloc[0] # similar/identical concept: Normalized Price with Base Value 1
```

**Drawback of Investment Multiple:** Doesn´t take account into investment Perioad. Meaningful only in conjunction with Investmend Period.
**Compound Annual Growth Rate (CAGR):** The (constant annual) rate of return that would be required for an investment to grow from its beginning balance to its ending balance, assuming the profits were reinvested at the end of each year of the investment´s lifespan.

```python
start=msft.index[0]
start
```

```python
end=msft.index[-1]
end
```

```python
# calculate the time difference (td)

td=end-start
td
```

```python
td_years=td.days/365.25
td_years
```

```python
# calculate the Compound Annual Growth Rate
cagr=multiple**(1/td_years)-1 #short version
cagr
```

```python
cagrII=(msft.Price.iloc[-1]/msft.Price.iloc[0])**(1/((msft.index[-1]-msft.index[0]).days/365.25))-1
cagrII
```

```python
(1+cagr)**td_years# alternative #2 to calculate multiple (cagr)
```

** -> CAGR can be used to compare Investments with different Investments horizon.**

# Compund Returns & Geometric Mean Return

```python
multiple=(msft.Returns+1).prod()
multiple
```

```python
n=msft.Returns.count()
n
```

```python
geo_mean=multiple**(1/n)-1 # Gemetric Mean return (daily)
geo_mean
```

```python
(geo_mean+1)**n
```

```python
mu=msft.Returns.mean() # arithmetic mean return
mu
```

**The arithmetic mean return is always greater than the gemoetric mean return... and less usefull**

```python
(mu+1)**n # calculate multiple? not possible with arithmetic mean!
```
