```python
compounding_policy=1

def future_values(pv,r,n,m=compounding_policy):
    if not m:
        m=1
    fv=pv*(1+r/m)**(n*m)
    return fv
```

```python
future_values(1000,0.04,5)
```

```python
compounding_policy = 1

def future_values(pv, r, n, m=compounding_policy):
    print(f"Defaultwert m = {m}")
    fv = pv * (1 + r/m) ** (n * m)
    return fv

# Funktionsaufruf 1 (ohne m - verwendet compounding_policy=1)
print(future_values(1000, 0.05, 10))

# Ändern von compounding_policy - aber Default in Funktion bleibt gleich!
compounding_policy = 12

# Funktionsaufruf 2 (ohne m - erwartet eigentlich m=12, aber Default ist noch 1)
print(future_values(1000, 0.05, 10))
```

```python
compounding_policy = 1

def future_values(pv, r, n, m=None):
    if m is None:
        m = compounding_policy
    print(f"Verwendeter Wert m = {m}")
    fv = pv * (1 + r/m) ** (n * m)
    return fv

# Funktionsaufruf 1 (ohne m - verwendet compounding_policy=1)
print(future_values(1000, 0.05, 10))

# Ändern von compounding_policy
compounding_policy = 12

# Funktionsaufruf 2 (ohne m - jetzt wird m=12 verwendet)
print(future_values(1000, 0.05, 10))

```
