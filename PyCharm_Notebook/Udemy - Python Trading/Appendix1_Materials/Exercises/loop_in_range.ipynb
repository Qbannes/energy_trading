```python
# Liste erstellen cf=casflows
cf=[100,10,20,50,30,25]
cf
```

```python
# Liste erstellen von 0-5
n=list(range(6))
n
```

```python
n=n[::-1]
n
```

```python
f=1.03
```

```python
FV=0
for i in range(6):
    FV+=cf[i]*f**n[i]
    print(FV)

# gegeben: f=1.03; n=[5,4,3,2,1,0]; cf[100,10,20,50,30,25]
# Ausgabe:
# 1) FV= 0+(100*1.03^5)
#    FV= 115.92740743
# 2) FV= 115.92740743+(10*1.03^4)
#    FV= 127.18249553000001
# 3) FV= 127.18249553000001+(20*1,03^3)
#    FV= 149.03703553000003
# 4) FV= 149.03703553000003+(50*1,03^2)
#    FV= 202.08203553
# 5) FV= 202.08203553+(30*1,03^1)
#    FV= 232.98203553000002
# 6) FV= 232.98203553000002+(25*1,03^0)
#    FV= 257.98203553000002
```

```python
# Wir können uns auch das Endergebnis direkt ausgeben lassen, indem wir die Printfunktion aus der Schleife rausnehemn:

FV=0
for i in range(6):
    FV+=cf[i]*f**n[i]
print(FV)

# Ausgabe: 257.98203553
```
