# DataFrame Operations

```python
import pandas as pd
```

```python
df = pd.DataFrame({'spa1':[1,2,3,4],
                   'spa2':[444,555,666,444],
                   'spa3':['abc','def','ghi','xyz']})
```

```python
df.head()
```

```python
df
```

```python
# einmalige Werte filtern

df['spa2'].unique()
```

```python
# ANzahl einmaliger Werte filtern

df['spa2'].nunique()
```

```python
# Die FUnktion agiert wie groupby() der spa2 und gibt die Anzahl der Werte aus

df['spa2'].value_counts()
```

```python
# neuen DataFrame vom bestehenden erstellen mit einer Bedingung (nehme alle Werte, die in der spa1 größer 2 sind in das neue DataFrame auf)
new_df= df[(df['spa1']>2)]
```

```python
new_df
```

```python
# Zweite Bedingung/Einschränkung (Nehme alle Werte aus spa2, die gleich 444 sind)

new_df= df[(df['spa1']>2)&(df['spa2']==444)]
```

```python
new_df
```

```python
# Funktion einer einfachen Multiplikation erstellen

def mal2(x):
    return x*2
```

```python
# Funktion mal2 auf spa2 anwenden mit .apply(mal2)

df['spa2'].apply(mal2)
```

```python
# Länge der Strings in spa3 ausgeben mit .apply(len)

df['spa3'].apply(len)
```

```python
# Lambda-Funktion anwenden und die Werte in spa2 mit 2 multiplizieren

df['spa2'].apply(lambda x: x*2)
```

```python
# mal 2

df['spa2']*2
```

```python
# Aggregationsfunktion

df['spa1'].sum()
```

```python
df['spa1'].min()
```

```python
df['spa1'].max()
```

```python
df.drop('spa1',axis=1)
```

```python
df
```

```python
# spa1 dauerhaft löschen mit del df['spa1]

del df['spa1']
```

```python
# Atribut Spalte ausgeben lassen

df.columns
```

```python
# Atribut Index ausgeben lassen

df.index
```

## NULL-Werte befüllen

```python
import numpy as np
```

```python
# DataFrame mit NULL-Werten erstellen

df = pd.DataFrame({'spa1':[1,2,3,np.nan],
                   'spa2':[np.nan,555,666,444],
                   'spa3':['abc','def','ghi','xyz']})
```

```python
df.head()
```

```python
# NULL-Werte befüllen

df.fillna(0)
```

## Pivotieren

```python
data = pd.DataFrame({'A':['foo','foo','foo','bar','bar','bar'],
                    'B':['one','one','two','two','one','one'],
                    'C':['x','y','x','y','x','y'],
                    'D':[1,3,2,5,4,1]})

df=pd.DataFrame(data)
```

```python
df
```

```python
df.pivot_table(values='D',index=['A','B'],columns='C')
```
