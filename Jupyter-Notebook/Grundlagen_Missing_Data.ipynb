# Missing Data

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({'A':[1,2,np.nan], 'B':[5,np.nan,np.nan],'C':[1,2,3]})
```

```python
df
```

## Eliminierungsverfahren
- Informationsverlust durch eliminieren der NaN-Werte (NaN = Not a Number)

```python
print(df.dropna()) # Es werden nur noch Zeilen angegeben, die keine NULL-Werte (NaN-Werte) enthält. Problem: Der Wert 2 in Zeile 1 ist auch weg
```

```python
print(df.dropna(axis=1)) # Spaltenauswahl die keine NULL-Werte enthält
```

```python
df.dropna(thresh=2) # alle Zeilen rausfiltern, die mindestens 2 Nullwerte enthält
```

```python
df.fillna(value='Füllwert')
```

```python
df.fillna(value=0)
```

```python
df['A'].fillna(value=df['A'].mean()) # Durchschnittswert der Spalte A
```

```python
df['A'].fillna(value=df['A'].sum()) # Durchschnittswert der Spalte A
```
