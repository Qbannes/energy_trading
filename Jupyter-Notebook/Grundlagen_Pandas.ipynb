```python
import numpy as np
import pandas as pd
```

```python
# 4 Inputdatensätze erstellen

labels = ['a','b','c']
liste = [10,20,30]
arr = np.array(liste)
dic = {'a':10,'b':20,'c':30}
```

```python
pd.Series(data=liste)
```

```python
pd.Series(data=liste, index=labels)
```

```python
# kürzere Schreibweise: Reihenfolge beachten: Daten (hier: die Liste) und dann Index (hier: die Labels) übergeben

pd.Series(liste, labels)
```

```python
# Da kein Index übergeben wird, wird von 0 inkrementiert (0,1,2,3,...)

pd.Series(arr)
```

```python
pd.Series(arr,labels)
```

```python
# Pandas Series erstellt

pd.Series(dic)
```

```python
# Pandas Series können Strings beinhalten

pd.Series(data=labels)
```

```python
# Funktionen übergeben

pd.Series([sum,print,len])
```

```python
# Index nutzen, um schnelle Suchen durchzuführen

ser1 = pd.Series([1,2,3,4],index=['usa','deutschland','russland','japan'])
```

```python
ser1
```

```python
ser2 = pd.Series([5,6,7,8],index=['usa','deutschland','italien','japan'])
```

```python
ser2
```

```python
ser2['usa']
```

# Operationen

```python
# Die Werte (Indizes), die Pandas nicht findet kann er nicht addieren und gibt keinen Wert aus (NaN) aus 

ser1 + ser2
```
