# **NumPy Operations**

```python
import numpy as np
```

```python
arr = np.arange(0,10)
```

```python
arr
```

```python
arr + arr
```

```python
arr * arr
```

```python
arr ** arr
```

```python
arr - arr
```

```python
# Warnung bei 0 / 0: Wert NaN wird zurückgegeben

arr / arr
```

```python
# Warnung: eine Zahl durch 0 ergibt inf (unendlich)

1/arr
```

```python
arr**3
```

```python
# quadratwurzel ziehen

np.sqrt(arr)
```

```python
# Exponenten
np.exp(arr)
```

```python
np.max(arr)
```

```python
arr.max()
```

```python
np.sin(arr)
```

```python
# Log 0 kann nicht berechnet werden und es wird inf (unendlich) ausgegeben 

np.log(arr)
```
