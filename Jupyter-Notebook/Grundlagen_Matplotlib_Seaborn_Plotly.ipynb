# Objektorientierte Methode

```python
# Leere Arbeitsfläche (af) erzeugen

af=plt.figure()
eigener_plot=af.add_axes([0.1,0.1,0.8,0.8]) # Liste mit Werten übergeben in der Klammer ([Position links unten(0.1),Postion unten(0.1), Breite der Arbeitsfläche(0.8),Höhe der Arbeitsfläche(0.8)])
eigener_plot.plot(x,y,'b')# verschiedene Daten übergeben
eigener_plot.set_xlabel('X Label')  # Labels setzen
eigener_plot.set_ylabel('Y Label') 
eigener_plot.set_title('Überschrift') # Überschrift setzen
```

```python
af=plt.figure()

axe1=af.add_axes([0.1,0.1,0.8,0.8])
axe2=af.add_axes([0.2,0.5,0.4,0.3])

axe1.plot(x,y,'b')
axe1.set_xlabel('X Label') 
axe1.set_ylabel('Y Label') 
axe1.set_title('groß')


axe2.plot(y,x,'r')
axe2.set_xlabel('X Label') 
axe2.set_ylabel('Y Label') 
axe2.set_title('klein')
```

```python
# Transparenz des Graphen ändern mit alpha=

diag,ax=plt.subplots()
ax.plot(x,x**2,color='blue',alpha=0.2)
```

```python
# An color können auch RGB Hex Codes übergeben werden 

diag,ax=plt.subplots()
ax.plot(x,x**2,color="#6B1FCF",alpha=1)
```

```python
# Linienbreite linewidth


diag,ax=plt.subplots()
ax.plot(x,x**2,color="#6B1FCF",alpha=1,linewidth=0.5)
```

```python
# Kürzere Schreibweise von linewidth


diag,ax=plt.subplots()
ax.plot(x,x**2,color="#6B1FCF",alpha=1,lw=0.5)
```

```python
# Linienstile (linestyle) (kurze Schreibweise ls)


diag,ax=plt.subplots()
ax.plot(x,x**2,color="#6B1FCF",alpha=1,linewidth=0.5,linestyle=':')
```

```python
# Linienstile (linestyle) (kurze Schreibweise ls)


diag,ax=plt.subplots()
ax.plot(x,x**2,color="#6B1FCF",alpha=1,linewidth=0.5,ls='dashed')
```

```python
# Marker


diag,ax=plt.subplots()
ax.plot(x,x**2,color="#6B1FCF",alpha=1,linewidth=0.5,ls='dashed', marker='x')
```

```python
# Markergröße


diag,ax=plt.subplots()
ax.plot(x,x**2,color="#6B1FCF",alpha=1,linewidth=0.5,ls='dashed', marker='x',markersize=10)
```

```python
# Markergröße kurz: ms


diag,ax=plt.subplots()
ax.plot(x,x**2,color="#6B1FCF",alpha=1,linewidth=0.5,ls='dashed', marker='x',ms=10)
```

```python
# versch. Markervariationen


diag,ax=plt.subplots()
ax.plot(x,x**2,color="#6B1FCF",alpha=1,linewidth=0.5,ls='dashed', marker='o',ms=20,markerfacecolor='yellow',markeredgewidth=3,markeredgecolor='green')
```

```python
# Axen limitieren (y)

diag,ax=plt.subplots()
ax.plot(x,x**2,color="#6B1FCF",alpha=1,linewidth=0.5,ls='dashed', marker='o',ms=20,markerfacecolor='yellow',markeredgewidth=3,markeredgecolor='green')
ax.set_ylim([0,5])



```

```python
# Axen limitieren (x)


diag,ax=plt.subplots()
ax.plot(x,x**2,color="#6B1FCF",alpha=1,linewidth=0.5,ls='dashed', marker='o',ms=20,markerfacecolor='yellow',markeredgewidth=3,markeredgecolor='green')
ax.set_ylim([0,20])
ax.set_xlim([0,5])
```
