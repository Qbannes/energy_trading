# Group By

```python
import pandas as pd
```

```python
# Data Frame erstelllen
data={'Firma':['GOOG','GOOG','MSFT','MSFT','FB','FB'], 'Person':['Sam','Charlie','Amy','Vanessa','Carl','Sarah'],'Sales':[200,12,340,124,243,350]}
```

```python
df=pd.DataFrame(data)
```

```python
print(df)
```

```python
df.groupby('Firma') # GroupBy-Object erhalten
```

```python
df.groupby('Firma')['Sales'].mean() # Durchschnitt der Spalte Sales berechnen
```

```python
# oder Druchschnitt der Spalten mit numerischen Werten berechnen. Hier: Spalte Sales

df.groupby('Firma').mean(numeric_only=True)
```

```python
# oder

df.groupby('Firma').mean('Sales')
```

```python
nach_firma=df.groupby('Firma')
```

```python
nach_firma.mean('Sales')
```

```python
nach_firma.std(numeric_only=True)
```

```python
nach_firma.min()
# oder
#nach_firma.min(numeric_only=True)
```

```python
nach_firma.max('Sales')
# oder
#nach_firma.ax(numeric_only=True)
```

```python
# Um auch die Spalte Person mit anzuzeigen, muss hier mit Index gearbeitet werden. 

# Den Index der Zeilen finden, in denen der maximale Sales pro Firma steht
idx = df.groupby('Firma')['Sales'].idxmin()
# idx = df.groupby('Firma')['Sales'].idxmax()

# Mit diesen Indizes die Zeilen aus dem DataFrame auswählen
result = df.loc[idx]

# Optional: Nach Firma sortieren und Index auf Firma setzen
# result_sorted = result.set_index('Firma').sort_index()

result_sorted
```
